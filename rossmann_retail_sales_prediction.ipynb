## ***3. Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# -----------------------------
# STEP 1: MERGING BOTH DATASETS
# -----------------------------
# 📌 Copying data to preserve raw versions
sales_df = Sales_df.copy()
store_df = Store_df.copy()

# 🧩 Merging both datasets using 'Store' as the key
df = pd.merge(sales_df, store_df, on = "Store", how = "left")


#==============================================================================================#


# -----------------------------
# STEP 2 : Handle unrealistic values
# ----------------------------
# 🧠 Replacing invalid competition years (before 1970) with NaN
df.loc[df['CompetitionOpenSinceYear'] < 1970, 'CompetitionOpenSinceYear'] = np.nan

# 🏫 Handling holidays and store info columns
df['SchoolHoliday'] = df['SchoolHoliday'].astype(int)


#==============================================================================================#



# -----------------------------
# STEP 3: Missing-value indicator flags
# -----------------------------
# 🚩 Creating missing-value indicator flags to identify records with null values after imputing for the same.
df['CompetitionDistance_NA'] = df['CompetitionDistance'].isna().astype(int)
df['CompetitionOpenSince_NA'] = df[['CompetitionOpenSinceMonth', 'CompetitionOpenSinceYear']].isna().any(axis=1).astype(int)
df['Promo2Since_NA'] = df[['Promo2SinceWeek', 'Promo2SinceYear']].isna().any(axis=1).astype(int)


#==============================================================================================#


# -----------------------------
# STEP 4 : Handle missing values
# -----------------------------
# 🎯 Handling promotion-related columns
df['Promo2SinceWeek'].fillna(0, inplace=True)
df['Promo2SinceYear'].fillna(0, inplace=True)
df['PromoInterval'].fillna('None', inplace=True)

# 🧹 Handling missing values for 'CompetitionDistance'
df['CompetitionDistance'] = df['CompetitionDistance'].fillna(df['CompetitionDistance'].median())

# 🧩 Handling missing values for competition opening details
df['CompetitionOpenSinceMonth'].fillna(0, inplace=True)
df['CompetitionOpenSinceYear'].fillna(0, inplace=True)



#==============================================================================================#


# -----------------------------
# STEP 5 : Feature extraction from date
# -----------------------------
# 🕒 Converting 'Date' column to datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')

# 📆 Extracting useful time-based features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month



#==============================================================================================#


# -----------------------------
# STEP 6 : Competition open duration
# -----------------------------
# 🕓 Creating 'CompetitionOpenMonths' — duration of competition presence
df['CompetitionOpenMonths'] = np.where(
    df['CompetitionOpenSince_NA'],
    0,  # Set to 0 if missing
    ((df['Year'] - df['CompetitionOpenSinceYear']) * 12 +
     (df['Month'] - df['CompetitionOpenSinceMonth']))
)

# Replacing negative values (for stores opened later) with 0
df['CompetitionOpenMonths'] = df['CompetitionOpenMonths'].apply(lambda x: x if x > 0 else 0)



#==============================================================================================#



# -----------------------------
# STEP 7 : Fixing data type issues
# -----------------------------
df['StateHoliday'] = df['StateHoliday'].astype('str').replace({'0': 'None'})
df['StoreType'] = df['StoreType'].astype('category')
df['Assortment'] = df['Assortment'].astype('category')
df['PromoInterval'] = df['PromoInterval'].astype('category')


#==============================================================================================#


# -----------------------------
# STEP 8 : Promo2 active (vectorized)
# -----------------------------
# 🧠 Creating 'IsPromo2Active' — checks if store's long-term promo is active
month_map = {'Jan':1, 'Feb':2, 'Mar':3, 'Apr':4, 'May':5, 'Jun':6,
             'Jul':7, 'Aug':8, 'Sept':9, 'Oct':10, 'Nov':11, 'Dec':12}

# Convert to string before applying function to avoid 'unhashable list' error
df['PromoMonths'] = df['PromoInterval'].astype(str).apply(
    lambda x: [month_map[m] for m in x.split(',')] if x != 'None' else []
)


df['IsPromo2Active'] = df.apply(
    lambda row: 1 if (row['Promo2'] == 1) and
                     (row['Month'] in row['PromoMonths']) and
                     (row['Year'] > row['Promo2SinceYear']) else 0, axis=1)



#==============================================================================================#



# -----------------------------
# STEP 9 : Business logic flags
# -----------------------------
# 💰 Creating 'SalesPerCustomer' — derived metric for customer efficiency
df['SalesPerCustomer'] = np.where(df['Customers'] > 0, df['Sales'] / df['Customers'], 0)

# 💰 Creating ZeroSalesWhileOpen to check anoamlies or Unproductive Day.
df['ZeroSalesWhileOpen'] = ((df['Sales'] == 0) & (df['Open'] == 1)).astype(int)



#==============================================================================================#



# -----------------------------
# STEP 10 : Handle outliers
# -----------------------------
# Cap competition distance at 99th percentile
cap_value = df['CompetitionDistance'].quantile(0.99)
df.loc[df['CompetitionDistance'] > cap_value, 'CompetitionDistance'] = cap_value



#==============================================================================================#



# -----------------------------
# STEP 11 : Final formatting
# -----------------------------
# 🚫 Dropping redundant and unnecessary columns
df.drop(columns=[
    'Promo2SinceYear',    # Used only for IsPromo2Active
    'PromoMonths',       # Temporary calculation column
    ], inplace=True)


# 🧾 Reordering columns for better readability
df = df[['Store', 'Date', 'Year', 'Month', 'DayOfWeek', 'Open', 'Promo', 'Promo2', 'IsPromo2Active',
         'Sales', 'Customers', 'SalesPerCustomer','ZeroSalesWhileOpen', 'StateHoliday', 'SchoolHoliday',
         'StoreType', 'Assortment', 'CompetitionDistance', 'CompetitionDistance_NA',
         'CompetitionOpenSinceMonth', 'CompetitionOpenSinceYear', 'CompetitionOpenSince_NA',
         'CompetitionOpenMonths', 'Promo2SinceWeek', 'Promo2Since_NA', 'PromoInterval']]


# Reset index and sort
df.sort_values(['Store', 'Date'], inplace=True)
df.reset_index(drop=True, inplace=True)


# ✅ Displaying final info
print("✅ Data Wrangling Completed Successfully!")
print(f"Final Dataset Shape: {df.shape}")
print("\nMissing Values After Wrangling:")
print(df.isnull().sum()[df.isnull().sum() > 0])
display(df.head(3))
````

### What all manipulations have you done and insights you found?

* During the data wrangling phase, multiple preprocessing and feature-engineering steps were performed to clean, correct, and enhance the dataset — making it ready for meaningful analysis and modeling.

* The manipulations focused on merging data, handling missing values, correcting data types, creating new features, and ensuring data consistency and reliability.


---

🔧 Manipulations Performed

1️⃣ Merging Datasets

* The two datasets — Sales_df (daily sales data) and Store_df (store information) — were merged on the common key Store.

* This merge allowed each sales record to include corresponding store-level details such as store type, assortment, and competition data.



---

2️⃣ Handling Unrealistic and Missing Values

* Unrealistic competition years (before 1975) were replaced with NaN to maintain data validity.

* Missing values in important fields like CompetitionDistance, CompetitionOpenSinceMonth/Year, and Promo2SinceWeek/Year were handled logically using median or zero imputation where applicable.

* Additional missing-value indicator columns (CompetitionDistance_NA, CompetitionOpenSince_NA, Promo2Since_NA) were created to capture the information about missingness — since even the absence of data can be predictive.



---

3️⃣ Data Type Corrections

* Columns such as StoreType, Assortment, and PromoInterval were converted to categorical data types to improve memory efficiency and interpretability.

* The StateHoliday column was kept as a categorical feature to allow one-hot encoding later during the modeling phase.



---

4️⃣ Feature Extraction from Date

* The Date column was transformed into datetime format, and new temporal features were extracted:
Year, Month, and DayOfWeek.

* These features will help identify seasonal trends, monthly variations, and weekday vs. weekend effects on sales.



---

5️⃣ Feature Engineering

* CompetitionOpenMonths → Calculated the number of months since a competitor store opened, set to 0 for missing or invalid entries.

* IsPromo2Active → A binary flag indicating if a store’s long-term promotion (Promo2) was active during the specific month, based on PromoInterval, Promo2SinceWeek, and Promo2SinceYear.

* SalesPerCustomer → Derived metric showing average customer spending, helping evaluate store-level performance efficiency.

* ZeroSalesWhileOpen → Business rule flag identifying anomalies where sales were 0 despite the store being open.



---

6️⃣ Outlier Treatment

* Extreme values in CompetitionDistance were capped at the 99th percentile to minimize the influence of outliers and maintain stable model behavior.



---

7️⃣ Sorting, Indexing, and Final Formatting

* The dataset was sorted by Store and Date to maintain chronological order.

* Redundant columns like Promo2SinceYear and intermediate helper variables were dropped for clarity.

* Index was reset to ensure a clean, analysis-ready DataFrame.




---

2. Some stores have long-standing competitors, indicating that mature markets may exhibit more stable sales.

4. Records with zero sales while open highlight potentially unproductive days or data inconsistencies that warrant deeper business review.

#### Chart 10 - Area Plot of Monthly Distribution.

````python
# Chart - 10 visualization code


# Features Used - Month

# ✅ Why this chart is important to include
# The Month column captures the temporal dimension of the data —
# which is essential for understanding seasonality in retail.
# Analyzing it helps identify:
#     Which months are busiest, indicating peak sales or promotion periods.
#     Low-activity months, which might need marketing push or inventory optimization.
#     Any missing or uneven data distribution across months (important for modeling).
#     The area plot adds visual depth, showing both trends and magnitudes over time,
#         perfect for highlighting seasonal variation.



#------------------------------------------------------------------------------------------------------#



# Creating a series of month-wise counts and ensuring months 1..12 are present
month_counts = df['Month'].value_counts().sort_index()
month_counts = month_counts.reindex(range(1,13), fill_value=0)  # Ensuring complete month index

# Creating the figure and defining its size
plt.figure(figsize=(10,6))

# Plotting the area chart using fill_between (x, y1)
# Passing x and y1 as positional arguments to satisfy matplotlib API
plt.fill_between(
    month_counts.index,            # x: month numbers 1..12
    month_counts.values,           # y1: counts for each month
    color='lightcoral',            # Choosing a soft red tone for retail theme
    alpha=0.7,                     # Adding transparency for aesthetic look
    linewidth=2,
    edgecolor='darkred'
)

# Overlaying a line for clearer trend visualization
plt.plot(
    month_counts.index,
    month_counts.values,
    marker='o',
    color='darkred',
    linewidth=1.8
)

# Adding title and axis labels
plt.title('Monthly Distribution of Records', fontsize=16, fontweight='bold', color='darkred')
plt.xlabel('Month (1 = Jan, 12 = Dec)', fontsize=12)
plt.ylabel('Number of Records', fontsize=12)

# Customizing ticks to show month numbers
plt.xticks(ticks=np.arange(1,13), labels=[str(m) for m in range(1,13)])

# Customizing grid and removing top/right borders
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.box(False)

# Displaying the plot
plt.show()

````

Data volume is highest in Jan-Jun (About 100K records each) and drops sharply after July (~60-65K).


This may reflect seasonality or partial data coverage and should be examined before modeling.

##### 1. Why did you pick the specific chart?

* I have used an area plot because it gives a clear and continuous visualization of how the number of records varies across different months.

* Since Month is a time-based feature, an area plot helps me spot seasonal activity patterns and check whether any month has missing or reduced data.

* It is also more visually appealing than a simple countplot for showing monthly volume trends.


##### 2. What is/are the insight(s) found from the chart?

* Exact month-wise counts (1 = Jan … 12 = Dec):

Jan = 103,694   Feb = 93,660     Mar = 103,695

Apr = 100,350   May = 103,695    Jun = 100,350

Jul = 98,115    Aug = 63,550     Sep = 61,500

Oct = 63,550    Nov = 61,500     Dec = 63,550


* The data shows that the first half of the year (Jan - Jun) has consistently higher counts (About 100K records each) compared to the second half (Aug - Dec) (~60-64K records each).

#### Chart 12 - PromoInterval (Months When Long-Term Promotions Are Active)

````python
# Chart - 12 visualization code


# Features Used - PromoInterval

# ✅ Why this chart is important to include ?
# The PromoInterval column defines which months (e.g., "Feb,May,Aug,Nov") a store’s long-term
# promotion (Promo2) is active.
# Analyzing this helps Rossmann understand:
#   Which months are most commonly used for recurring promotions.
#   Whether promotional activity is evenly spread or clustered in specific months.
#   Seasonal promotional strategy — for instance, heavier promotions in mid-year or near holidays.
#   This chart provides clear visibility into Rossmann’s annual promotion schedule.


#--------------------------------------------------------------------------------------------------------#

# Keeping only non-null PromoInterval values
promo_df = df.dropna(subset=['PromoInterval']).copy()

# Splitting comma-separated months into lists efficiently
promo_df['PromoInterval'] = promo_df['PromoInterval'].str.split(',')

# Using explode() to create one row per month (much faster than sum())
promo_month_counts = (
    promo_df['PromoInterval']
    .explode()                              # Expanding month list into separate rows
    .value_counts()                         # Counting occurrences of each month
    .sort_index()                           # Sorting months alphabetically
)

# Creating the figure and defining its size
plt.figure(figsize=(9,5))

# Plotting bar chart for frequency of months in PromoInterval
sns.barplot(
    x=promo_month_counts.index,
    y=promo_month_counts.values,
    palette='crest',
    edgecolor='black'
)

# Adding title and labels
plt.title('Frequency of Months in PromoInterval', fontsize=15, fontweight='bold', color='darkgreen')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Stores Running Promotions', fontsize=12)

# Adding count labels on bars
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%d', label_type='edge', fontsize=10)

# Customizing grid and borders
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.box(False)

# Displaying the plot
plt.show()
````

**Summary -**

PromoInterval months Jan, Apr, Jul, Oct (≈ 293 K each) dominate — reflecting Rossmann's quarterly promotion cycle.


Around 508 K records have “No PromoInterval”, meaning no long-term promo was active for those periods.


##### 1. Why did you pick the specific chart?

* I have used a countplot because PromoInterval is a categorical column that lists the months in which recurring, long-term promotions (Promo2) are active.

* The countplot helps me clearly visualize which months are most frequently used for continuous promotions, giving an overview of the marketing calendar pattern adopted by Rossmann stores.

* This insight is valuable for identifying promotional clusters and understanding if the campaigns are evenly spread or concentrated in certain months.

##### 2. What is/are the insight(s) found from the chart?

* The chart shows that the months January, April, July, and October have the highest frequency (≈ 293 K each) — these are the four most common intervals for recurring promotions.

* This confirms that Rossmann’s long-term promotions follow a quarterly pattern, launched roughly every three months.

* The large count under “None” (≈ 508 K) indicates that many records correspond to stores or periods without any long-term promo active, which aligns with about half the stores not participating in Promo2.

Are there any insights that lead to negative growth? Justify with specific reason.

**Positive business impacts / what to do:**

##### Chart 17 - Promo vs Sales

````python
# Chart - 17 visualization code

# Features Used - Promo, Sales

# ✅ Why this chart is important to include?
# I am creating this chart to visually compare sales performance between promotional and non-promotional days.
# This helps in understanding how promotions influence daily sales levels and overall business revenue patterns.
# It also helps confirm if promotional campaigns are truly effective or just causing short-term spikes.

# ---------------------------------------------------------------------------------------------------------- #

# Setting visual style for professional presentation
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Plotting the boxplot to compare sales distribution between promo and non-promo days
sns.boxplot(
    data=df,
    x='Promo',
    y='Sales',
    palette={'0': '#2E86C1', '1': '#E67E22'},
    showfliers=False
)

# Adding median annotations
medians = df.groupby('Promo')['Sales'].median().values
for i, median in enumerate(medians):
    plt.text(i, median + 300, f"Median = {int(median):,}",
             ha='center', fontweight='semibold', color='black')

# Adding titles and labels
plt.title('Distribution of Sales during Promo vs Non-Promo Days', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Promo (0 = No Promo, 1 = Promo)', fontsize=12)
plt.ylabel('Sales Amount', fontsize=12)

# Customizing gridlines and layout
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

# Displaying the plot
plt.show()
````

Non-Promo Days: 629,129 (≈ 61.8%)

Median (Promo = 0): 4,622

Median (Promo = 1): 7,553 → ≈ +63.4% uplift

Mean (Promo = 0): 4,406

Mean (Promo = 1): 7,991 → ≈ +81.4% uplift

The chart clearly shows that stores experience a substantial increase in sales during promotion periods.

* Since promotions are key business levers, analyzing their impact on sales distribution is critical for campaign strategy and revenue forecasting.


###### 2. What is/are the insight(s) found from the chart?

* The IQR (spread) during promo days is wider, indicating that some promotions perform exceptionally well while others have moderate results.

* The median sales during non-promo days (₹4,622) are considerably lower than during promo days (₹7,553), confirming that promotions generally help increase sales volume.

* Overall, the chart confirms that promotions have a positive impact on sales performance but with variable outcomes across stores or time periods.

````python
# Chart - 19 visualization code

# Features Used - Month, Sales

# ✅ Why this chart is important to include ?
# This chart helps identify monthly sales trends and seasonal patterns across the year.
# Understanding which months drive higher or lower sales is crucial for inventory planning,
# staffing, and timing promotional campaigns for maximum impact.

#----------------------------------------------------------------------------------------------------------#

# Grouping data by Month to calculate average sales per month
monthly_sales = df.groupby('Month')['Sales'].mean().reset_index()

# Sorting by Month to ensure correct order (if not already sorted)
monthly_sales = monthly_sales.sort_values(by='Month')

# Setting the visual style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# ✅ Corrected area chart syntax (fill between baseline y=0 and actual sales)
plt.fill_between(
    x=monthly_sales['Month'],
    y1=monthly_sales['Sales'],
    y2=0,                       # baseline
    color='skyblue',
    alpha=0.4,
    linewidth=2,
    edgecolor='blue'
)

# Adding a smooth line on top for clarity
sns.lineplot(
    data=monthly_sales,
    x='Month',
    y='Sales',
    color='blue',
    linewidth=2.5
)

# Titles and labels
plt.title('Monthly Average Sales Trend', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Sales', fontsize=12)

# Customizing x-axis labels
plt.xticks(
    ticks=range(1,13),
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)

# Gridlines and aesthetics
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.box(False)

# Display the plot
plt.show()
````

##### Chart 23 - Impact of Competition Distance on Sales (Binned Line Chart)

````python
# Chart - 23 Visualisation Code

# Features Used - Sales, CompetitionDistance

# ✅ Why this chart is important to include?
# This chart helps analyze how average sales behave as the distance to the nearest
# competitor increases. By binning competition distance, we reduce noise from
# individual stores and clearly observe overall sales trends across proximity levels.
# This makes it easier to understand whether nearby competition negatively impacts sales.

# ---------------------------------------------------------------------------------------------------------- #

# Setting Seaborn style for professional aesthetics
sns.set(style='whitegrid')

# Creating competition distance bins
distance_bins = [0, 500, 1000, 3000, 10000, df['CompetitionDistance'].max()]
distance_labels = [
    'Very Close (<500m)',
    'Close (500m–1km)',
    'Moderate (1–3km)',
    'Far (3–10km)',
    'Very Far (>10km)'
]

# Assigning bins to CompetitionDistance
df['CompetitionDistanceBin'] = pd.cut(
    df['CompetitionDistance'],
    bins=distance_bins,
    labels=distance_labels,
    include_lowest=True
)

# Calculating average sales per competition distance bin
binned_sales = (
    df.groupby('CompetitionDistanceBin', observed=True)['Sales']
      .mean()
      .reset_index()
)

# Creating the figure
plt.figure(figsize=(10, 6))

# Line plot to show sales trend across competition distance bins
sns.lineplot(
    data=binned_sales,
    x='CompetitionDistanceBin',
    y='Sales',
    marker='o',
    linewidth=2,
    color='steelblue'
)

# Adding value labels on data points
for index, row in binned_sales.iterrows():
    plt.text(
        index,
        row['Sales'],
        f"{int(row['Sales'])}",
        ha='center',
        va='bottom',
        fontsize=10
    )

# Chart title and labels
plt.title(
    'Average Sales Across Competition Distance Buckets',
    fontsize=16,
    fontweight='bold'
)
plt.xlabel('Competition Distance', fontsize=12)
plt.ylabel('Average Sales', fontsize=12)

# Improving grid and layout
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

# Displaying the chart
plt.show()
````

**Summary -**

* This chart shows how average sales vary across different competition distance buckets.

* Sales are highest when competitors are very close   (< 500m), drop sharply when competitors are slightly farther away (500m - 1km), and then stabilize across moderate to very far distances.

* Overall, the relationship between competition distance and sales is non-linear, indicating that proximity alone does not consistently suppress sales.

###### 1. Why did you pick the specific chart?

* The binned line chart reduces noise from millions of individual observations.

* It clearly captures macro-level sales trends across competition distance ranges.

* Compared to regression/scatter plots, this chart is faster to compute and easier to interpret, making it ideal for EDA.

###### 2. What is/are the insight(s) found from the chart?

* Stores with very close competitors (< 500m) have the highest average sales (~6307).

* Sales drop significantly when competition is at a moderate proximity (500m–1km).

* Beyond 1km, sales remain relatively stable, fluctuating slightly but without a strong upward or downward trend.

* This suggests that store location quality and demand density may outweigh pure competition effects.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.


**✅ Positive business impact / opportunities:**

* Businesses should not avoid opening stores near competitors by default.

* High sales near close competition may indicate high-demand zones, making such areas strategically valuable.

* Expansion decisions can focus more on market demand rather than fear of nearby competitors.

**⚠️ Potential negative growth / risks:**

* The sharp decline in sales from < 500m to 500m–1km indicates a potential cannibalization zone.

* Stores located just outside high-demand clusters may face reduced visibility and footfall, leading to lower sales.

* This highlights a risk zone where competition exists but demand density may not be sufficient to sustain strong performance.

**👉 Actionable takeaway:**

* If entering a competitive area, it is better to be very close (core demand zone) rather than moderately distant.


##### Chart 24 - CompetitionOpenMonths vs Sales

````python
# Chart - 24 Visualisation Code

# Features Used - CompetitionOpenMonths, Sales

# ✅ Why this chart is important to include?
# I am creating this chart to measure whether the age of the nearest competitor (in months)
# is associated with a change in average store Sales. This tests whether older competition
# has a sustained impact or whether new competition causes more short-term disruption.

# ---------------------------------------------------------------------------------------------------------- #

# Preparing the series: exclude 0 (missing / no-competition flag) because 0 does not mean '0 months of competition' in business terms
comp_mask = df['CompetitionOpenMonths'] > 0
comp_df = df.loc[comp_mask, ['CompetitionOpenMonths', 'Sales']].copy()

# If there are no records with CompetitionOpenMonths > 0, this will safely handle the case
if comp_df.empty:
    print("No records with CompetitionOpenMonths > 0. Check CompetitionOpenMonths creation.")
else:
    # Aggregating median sales by exact months-since-competition
    monthly_comp = (
        comp_df.groupby('CompetitionOpenMonths', as_index=True)['Sales']
        .median()
        .rename('median_sales')
        .reset_index()
        .sort_values('CompetitionOpenMonths')
    )

    # Smoothing the median series with a rolling median to reduce noise (window = 5 months)
    monthly_comp['rolling_median'] = monthly_comp['median_sales'].rolling(window=5, center=True, min_periods=1).median()

    # To avoid ultra-long x axes (very rare extremely large months), capping x for plotting clarity:
    max_display_month = 120   # showing first 10 years for clarity; values beyond will still be considered in stats
    monthly_comp_plot = monthly_comp[monthly_comp['CompetitionOpenMonths'] <= max_display_month]

    # Count of zero / NA-flagged competition records (for annotation)
    zero_comp_count = df.loc[df['CompetitionOpenMonths'] == 0].shape[0]
    total_count = df.shape[0]

    # Plotting
    sns.set(style='whitegrid')
    plt.figure(figsize=(11,6))

    # Scatter of raw medians (light)
    plt.scatter(
        monthly_comp_plot['CompetitionOpenMonths'],
        monthly_comp_plot['median_sales'],
        s=30,
        alpha=0.6,
        color='steelblue',
        label='Median sales (per month bin)'
    )

    # Smoothed rolling median line
    plt.plot(
        monthly_comp_plot['CompetitionOpenMonths'],
        monthly_comp_plot['rolling_median'],
        color='#E67E22',
        linewidth=2.5,
        label='Smoothed median (rolling window=5)'
    )

    # Highlight vertical line at 12, 24, 60 months for reference (1yr, 2yr, 5yr)
    for m in [12, 24, 60]:
        plt.axvline(m, color='grey', linestyle='--', alpha=0.4)

    plt.title('Impact of Competitor Age (months since opened) on Median Store Sales', fontsize=16, fontweight='bold')
    plt.xlabel('Months since nearest competitor opened', fontsize=12)
    plt.ylabel('Median Daily Sales', fontsize=12)
    plt.xlim(0, monthly_comp_plot['CompetitionOpenMonths'].max() + 2)
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    plt.legend()

    # Annotating count of missing/zero competition months
    plt.gca().text(
        0.98, 0.02,
        f'Zero/NA competition records: {zero_comp_count:,} / {total_count:,}',
        ha='right', va='bottom', transform=plt.gca().transAxes,
        fontsize=10, bbox=dict(facecolor='white', alpha=0.6, edgecolor='none')
    )

    plt.tight_layout()
    plt.show()
````

**Summary -**

* Data excludes ~41% records (415,219 / 1,017,209) where competition data was missing (i.e., CompetitionOpenMonths = 0).

* Median daily sales hover around ₹5,800–6,200 during the first 40 months after a competitor opens.

* Between months 40–70, sales remain moderately stable but start declining sharply after ~70 months (≈ 6 years).

* After 6 years, sales drop close to ₹5,400–5,500, indicating mature competition impact.

* Interestingly, a slight rebound occurs between 90–110 months, possibly due to market adaptation or competitor closure.

* Overall, older competitor presence appears to dampen store sales after several years.


###### 1. Why did you pick the specific chart?

* I used a scatter + rolling median line chart because this combination effectively captures long-term, nonlinear patterns between competitor age and store performance.

* A simple correlation or regression would miss these temporal shifts and nonlinear effects.

* By smoothing medians, the chart shows whether competition effects are short-term shocks or long-term declines.

* This visualization is critical for business teams to understand how the “age of competition” shapes store sustainability over years.

###### 2. What is/are the insight(s) found from the chart?

* Initial stability (0–40 months): Sales remain consistent soon after competition opens — stores may retain loyal customers initially.

* Gradual decline (40–70 months): Sales start declining as competitors gain market foothold and customers start shifting.

* Significant drop (~70–85 months): The most visible fall — median sales drop by nearly ₹700, suggesting stronger competitor capture.

* Recovery trend (90–110 months): A small upward movement implies market rebalancing — possibly from adaptation, local marketing, or competitor exit.

* Plateau beyond 110 months: Sales stabilize again around ₹5,700.

* Overall, competitor impact grows over time, but the decline may stabilize after stores adapt or weak competitors exit.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.


**✅ Positive business impacts / What to do:**

* Strategy planning: New store openings should focus on areas where competitors are newly established (under 3 years) — minimal impact expected.

* Retention programs: For stores older than 6 years with mature nearby competitors, launch loyalty campaigns or local offers to reclaim customer base.

* Market adaptation: Recognize the sales dip period (~70 months) as a critical intervention window for management action.

* Long-term forecasting: Insights can help improve time-based competition modeling in future demand forecasts.


**⚠️ Negative growth / Risks:**

* If this trend continues, resource reallocation or format revamp (store type, assortment) might be needed for recovery.

##### Chart 27 - Effect of Daily Promo on Customers

````python
# Chart 27 - Visualization Code

# Features Used - Promo, Customers

# ✅ Why this chart is important to include?
# This chart helps me understand how promotions impact store footfall.
# If promotions are effective, I should see higher customer counts during promo periods.
# It’s useful for checking whether promotions drive more people into the store,
# or if they mostly affect existing customer spending (seen in sales data).

# ---------------------------------------------------------------------------------------------------------- #

# Setting consistent style
sns.set(style='whitegrid')

# Defining correct logical order for Promo variable
order = [0, 1]  # 0 = No Promo, 1 = Promo Active

# Creating the figure
plt.figure(figsize=(8,6))

# Plotting boxplot
sns.boxplot(
    data=df,
    x='Promo',
    y='Customers',
    order=order,
    palette=['#81D4FA', '#FFB74D'],
    width=0.5
)

# Adding titles and axis labels
plt.title('Effect of Promotions on Customer Footfall', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Promo (0 = No, 1 = Yes)', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

# Adding median annotations for each category
medians = df.groupby('Promo')['Customers'].median()
for i, val in enumerate(order):
    median_val = medians[val]
    plt.text(i, median_val + 100, f'Median = {int(median_val)}',
             ha='center', fontsize=10, fontweight='bold', color='black')

# Customizing visuals
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.box(False)

# Displaying the chart
plt.show()
````

##### Chart 29 - Monthwise Traffic Pattern

````python
# Chart 29 - Visualization Code
# Features Used - Month, Customers

# ✅ Why this chart is important to include?
# This chart helps me identify monthly patterns in customer footfall across all stores.
# It shows whether there are seasonal peaks or dips in customer visits,
# which is valuable for planning staffing, promotions, and inventory during high-traffic periods.

# ---------------------------------------------------------------------------------------------------------- #

# Grouping data by month to get average number of customers
monthly_customers = (
    df.groupby('Month')['Customers']
    .mean()
    .reset_index()
    .sort_values('Month')
)

# Setting consistent style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Plotting the filled area to show variation in customer count over months
plt.fill_between(
    x=monthly_customers['Month'],
    y1=monthly_customers['Customers'],
    y2=0,
    color='skyblue',
    alpha=0.4,
    linewidth=2,
    edgecolor='steelblue'
)

# Adding a smooth line for clarity
sns.lineplot(
    data=monthly_customers,
    x='Month',
    y='Customers',
    color='steelblue',
    linewidth=2.5,
    marker='o'
)

# Chart title and labels
plt.title('Monthly Trend of Average Customer Footfall', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Number of Customers', fontsize=12)

# Customizing x-axis labels
plt.xticks(
    ticks=range(1,13),
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)

# Adding gridlines
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.box(False)

# Displaying the plot
plt.show()
````

**Summary -**

* The average number of customers per month stays fairly stable throughout the year, between 600–670 on average.

* A slight dip appears around May and August, while July and December show small peaks.

* This indicates that customer visits don’t fluctuate dramatically by month — suggesting consistent year-round engagement.

###### 1. Why did you pick the specific chart?

* I wanted to understand how customer traffic changes month to month and whether there’s any strong seasonality in visits.

* An area plot clearly highlights the overall volume and flow of customer activity, making it easy to spot small ups and downs across the year.

* It’s also visually consistent with my earlier “Month vs Sales” chart, helping me connect sales patterns with actual store footfall.


###### 2. What is/are the insight(s) found from the chart?

* The chart shows that customer footfall remains steady all year, with no extreme spikes or drops.

* Slight increases are seen in July and December, likely due to summer holidays and the end-of-year shopping period.

* May and August show small declines — possibly vacation months when regular shopping frequency drops a bit.

* Overall, this means Rossmann enjoys a stable customer base, which is a great sign for operational predictability.


###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Positive business impact:**

* The stability of customer traffic suggests the brand has a loyal, repeat customer base, not just seasonal buyers.

* Managers can plan consistent inventory and staffing levels across months without major seasonal stress.

* The minor peaks (July and December) can be used as promo opportunities to capture slightly higher demand.


**Possible risks / negative impact:**

* The small dips in May and August could indicate vacation-driven slowdowns, which can be offset with light promotional campaigns.

##### Chart 30 - DayOfWeek vs Customers

````python
# Chart 30 - Visualization Code

# Features Used - DayOfWeek, Customers

# ✅ Why this chart is important to include?
# This chart helps me see on which days customers visit stores the most or the least.
# It’s useful for optimizing weekly staffing, scheduling promotions, and managing inventory restocks.
# Understanding weekday footfall patterns also helps improve store efficiency and sales strategy.

# ---------------------------------------------------------------------------------------------------------- #

# Grouping data by DayOfWeek and calculating average customer count
dow_customers = (
    df.groupby('DayOfWeek')['Customers']
    .mean()
    .reset_index()
    .sort_values('DayOfWeek')
)

# Setting consistent plot style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Plotting the barplot for average customers by day of the week
sns.barplot(
    data=dow_customers,
    x='DayOfWeek',
    y='Customers',
    palette='crest'
)

# Adding chart titles and labels
plt.title('Average Customer Footfall by Day of the Week', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Day of the Week (1 = Monday, 7 = Sunday)', fontsize=12)
plt.ylabel('Average Number of Customers', fontsize=12)

# Annotating values on top of bars
for index, row in dow_customers.iterrows():
    plt.text(row['DayOfWeek'] - 1, row['Customers'] + 20, f"{int(row['Customers'])}",
             ha='center', fontsize=10, fontweight='semibold', color='black')

# Customizing gridlines
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.box(False)

# Displaying the chart
plt.show()
````

##### Chart 32 - Customer Footfall Distribution Across Assortment Types

````python
# Chart 32 - Visualization Code

# Features Used: Assortment, Customers

# ✅ Why this chart is important to include?
# This chart is helping to understand how different assortment types
# are influencing customer footfall across stores.
# By visualizing the distribution of customers for each assortment category,
# this chart is highlighting whether broader assortments are consistently
# attracting higher customer traffic or if the impact is marginal.
# Using a strip plot with jitter is improving performance while still
# preserving the distribution insight needed for analysis.

# ---------------------------------------------------------------------------------------------------------- #

# Setting consistent visual style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(9,6))

# Plotting strip plot with jitter for faster rendering
sns.stripplot(
    data=df.sample(4000, random_state=42),   # Reducing sample size for speed
    x='Assortment',
    y='Customers',
    palette=['#64B5F6', '#81C784', '#FFB74D'],
    jitter=0.25,                              # Adding jitter to show spread
    size=3,
    alpha=0.6
)

# Adding chart title and labels
plt.title(
    'Customer Footfall Distribution Across Assortment Types',
    fontsize=16,
    fontweight='bold'
)
plt.xlabel('Assortment Type', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

# Adding gridlines for readability
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.box(False)

# Displaying the chart
plt.tight_layout()
plt.show()
````

**Summary -**

* The chart is showing a wide spread of customer footfall across all three assortment types (a, b, c), indicating significant variability at the store level.

* Assortment type c is visibly having a higher concentration of stores with larger customer counts, especially in the upper range (above 1500 customers).

* Assortment type a is showing a dense cluster in the mid-range (around 400–1200 customers), with fewer extreme high-footfall stores.

* Assortment type b is having the sparsest data points, which aligns with fewer stores belonging to this assortment type.

* Zero-customer observations are still visible across all assortment types, reinforcing the importance of earlier “Zero Sales While Open” analysis.

###### 1. Why did you pick the specific chart?

* I picked this chart to visually compare how customer footfall is distributed across different assortment strategies at the store level.

* I intentionally chose a strip plot with jitter instead of a swarm plot because swarm plots were taking excessive time to render on this dataset.

* The strip plot is allowing faster execution while still preserving the core objective of showing dispersion, clustering, and outliers.

* This chart is helping to evaluate whether richer assortments consistently translate into higher customer traffic.

###### 2. What is/are the insight(s) found from the chart?

* Stores with assortment type c tend to attract higher customer volumes more frequently than the other assortment types.

* Assortment type a appears to perform steadily but is less likely to produce very high-footfall stores.

* Assortment type b is both limited in count and inconsistent in customer attraction.

* The presence of many overlapping points suggests that assortment alone is not the only driver of customer traffic and must be analysed alongside other factors such as location, promotions, and competition.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Positive Business Impact:**

* The faster-performing visualization ensures smoother analysis workflows and better productivity.

##### Chart 33 - ZeroSalesWhileOpen Anomaly or Poor Performance ?

````python
# Chart - 33 Visualisation Code

# Features Used - Customers, ZeroSalesWhileOpen

# ✅ Why this chart is important to include?
# This chart helps me understand how customer footfall behaves on days where the store was open
# but sales recorded as zero. If customer count is high but sales are zero, it's a strong indicator
# of operational or data recording issues. If customer count is ~0 even when 'Open' shows 1,
# it suggests incorrect labeling or exceptional cases.
# This is crucial for identifying anomalies before modeling.

# ------------------------------------------------------------------------------------------------ #

# Setting plot style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Plotting boxplot
sns.boxplot(
    data=df,
    x='ZeroSalesWhileOpen',
    y='Customers',
    palette=['skyblue','salmon']
)

# Adding median labels manually
medians = df.groupby('ZeroSalesWhileOpen')['Customers'].median()
positions = range(len(medians))

for pos, median_val in zip(positions, medians):
    plt.text(
        pos, median_val + 20,               # position slightly above median line
        f"Median = {int(median_val)}",      # text
        horizontalalignment='center',
        fontweight='bold',
        color='black'
    )

# Adding titles and labels
plt.title("Impact of Zero Sales While Store Was Open on Customer Footfall",
          fontsize=16, fontweight='bold')
plt.xlabel("Zero Sales While Open (0 = No, 1 = Yes)", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

# Display the plot
plt.show()
````

**Summary -**

* When ZeroSalesWhileOpen = 0, the typical customer footfall has a median of 609 customers, showing normal store activity.

* When ZeroSalesWhileOpen = 1, the median customer count drops to 0, meaning these events occur only when customer footfall was absent or extremely low.

* The pattern strongly suggests that - zero-sales-while-open cases are either data labeling issues, incorrect open flag, or rare operational exceptions, not actual lost-sales events.

###### 1. Why did you pick the specific chart?

* I chose this chart because ZeroSalesWhileOpen is one of the most important anomaly-flag features created during data wrangling.

* To understand whether these events are due to genuine zero customer visits or due to operational failures, I need to compare customer footfall between normal days and anomaly days.

* A boxplot clearly shows the difference in medians and the spread, making it easy to validate the reliability of the ZeroSalesWhileOpen flag.

###### 2. What is/are the insight(s) found from the chart?

* For normal days (ZeroSalesWhileOpen = 0), stores receive a median of 609 customers, with a wide spread up to even 7000+ customers.

* For anomaly days (ZeroSalesWhileOpen = 1), the median customer count is 0, and almost all values are zero or extremely close to zero.

* This tells me that these events are not actually “sales failure despite customers being present”.
Rather, they look like:

    * Wrongly tagged open status,

    * Stores marked as open even though they were practically not operating,

    * Or extremely rare edge cases with truly no footfall.


* There is no evidence of high customer count combined with zero sales — which means no strong sign of POS failure or operational breakdown.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Positive business impact :**

* This analysis confirms that ZeroSalesWhileOpen does not represent revenue loss events — the stores had little to no customers anyway.

* I can treat most of these records as harmless anomalies rather than true operational failures.

* During modeling, I can safely keep this feature or even remove these records without worrying about losing meaningful business information.

* Understanding this prevents me from falsely assuming revenue leakage or system issues.


**Negative growth / risk signals:**

* Although the count of ZeroSalesWhileOpen is extremely small, their presence suggests inconsistent data labeling of store open/close status.

* If not cleaned, these anomalies could add noise to the forecasting model and slightly affect accuracy.

* Ensuring correct open/closed labels would improve the dataset's long-term reliability.

##### Chart 35 - Monthly Sales Per Customer

````python
# Chart 35 - Visualization Code
# Features Used - Month, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart helps me understand how average customer spending changes across months.
# If SalesPerCustomer increases or decreases consistently in certain months,
# it indicates seasonal shifts in how much each customer is willing to spend.
# This is useful for planning promotions, inventory, and seasonal strategies.

# ---------------------------------------------------------------------------------------------------------- #

# Grouping by month to calculate average SalesPerCustomer
monthly_spc = (
    df.groupby('Month')['SalesPerCustomer']
    .mean()
    .reset_index()
    .sort_values('Month')
)

# Setting consistent chart style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Line plot showing month-wise changes
sns.lineplot(
    data=monthly_spc,
    x='Month',
    y='SalesPerCustomer',
    color='darkgreen',
    linewidth=2.5,
    marker='o'
)

# Adding filled area to improve readability
plt.fill_between(
    monthly_spc['Month'],
    monthly_spc['SalesPerCustomer'],
    alpha=0.2,
    color='lightgreen'
)

# Chart Title and Labels
plt.title('Monthly Trend of Average Sales per Customer', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Sales per Customer', fontsize=12)

# Month labels
plt.xticks(
    ticks=range(1,13),
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)

# Final styling
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.box(False)

# Show plot
plt.show()
````

**Summary -**

* SalesPerCustomer stays fairly stable throughout the year, mostly between 7.5 and 8.1.

* Slight peaks occur in February, June, and November.

* Noticeable dips occur around May and October, where customers spend a little less on average.

* Overall, the average value remains steady, suggesting consistent customer spending behavior despite monthly variations in traffic.

###### 1. Why did you pick the specific chart?

* I chose this chart because it helps me understand whether monthly spending patterns change significantly throughout the year.

* Earlier charts showed that customer footfall stays almost stable, but this chart tells me if the amount each customer spends also remains stable or if there are seasonal peaks.

* A line plot smoothly highlights month-to-month changes and makes spotting subtle seasonal spending trends easier.

###### 2. What is/are the insight(s) found from the chart?

* Customers maintain a consistent spending pattern across the year with only mild fluctuations.

* The highest spending appears in February (≈8.0+), suggesting early-year campaigns or seasonal shopping influence spending.

* A dip occurs in May, where average SalesPerCustomer drops to around 7.4 - 7.5, indicating a mild slowdown in customer buying behavior.

* Another drop in October suggests customers may be delaying purchases until November or December promotional periods.

* Peaks in June and November hint that customers may take advantage of mid-year or year-end offers.

Are there any insights that lead to negative growth? Justify with specific reason.

**✅ Positive business impact:**

##### Chart 36 -  DayOfWeek vs SalesPerCustomer

````python
# Chart 36 - Visualization Code
# Features Used - DayOfWeek, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart helps me understand how average customer spending varies across the week.
# Even if customer footfall changes by day, the amount each customer spends may show a different pattern.
# Knowing which days customers spend more (or less) is useful for planning weekday promotions,
# staffing, and revenue forecasting.

# ---------------------------------------------------------------------------------------------------------- #

# Grouping by DayOfWeek to calculate average SalesPerCustomer
dow_spc = (
    df.groupby('DayOfWeek')['SalesPerCustomer']
    .mean()
    .reset_index()
    .sort_values('DayOfWeek')
)

# Setting consistent plot style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Bar plot showing average spending per customer across weekdays
sns.barplot(
    data=dow_spc,
    x='DayOfWeek',
    y='SalesPerCustomer',
    palette='viridis'
)

# Adding chart title and axis labels
plt.title('Average Sales per Customer by Day of the Week', fontsize=16, fontweight='bold', color='black')
plt.xlabel('Day of the Week (1 = Monday, 7 = Sunday)', fontsize=12)
plt.ylabel('Average Sales per Customer', fontsize=12)

# Adding annotations on top of bars
for index, row in dow_spc.iterrows():
    plt.text(
        row['DayOfWeek'] - 1,
        row['SalesPerCustomer'] + 0.1,
        f"{round(row['SalesPerCustomer'], 2)}",
        ha='center',
        fontsize=10,
        fontweight='semibold'
    )

# Customizing grid
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.box(False)

# Displaying the chart
plt.show()
````

````python
# Chart 37 - Visualization Code
# Features Used - IsPromo2Active, SalesPerCustomer

# ✅ Why this chart is important?
# This chart helps me analyze whether long-term promotions (Promo2)
# influence how much customers spend per visit.
# Promo2 is a recurring promotion active in specific months,
# so this visualization shows if it increases basket size or not.

# ---------------------------------------------------------------------------------------------------------- #

# Setting visual style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(8,6))

# Plotting boxplot for SalesPerCustomer vs Promo2 activity
sns.boxplot(
    data=df,
    x='IsPromo2Active',
    y='SalesPerCustomer',
    palette=['#AED6F1', '#F9E79F'],
    width=0.5
)

# Adding title and labels
plt.title('Impact of Long-Term Promotion (Promo2) on Sales per Customer',
          fontsize=16, fontweight='bold', color='black')
plt.xlabel('Is Promo2 Active? (0 = No, 1 = Yes)', fontsize=12)
plt.ylabel('Average Sales per Customer', fontsize=12)

# Adding median annotations
medians = df.groupby('IsPromo2Active')['SalesPerCustomer'].median()

for i, val in enumerate([0,1]):
    plt.text(
        i,
        medians[val] + 0.4,
        f"Median = {round(medians[val], 2)}",
        ha='center',
        fontsize=10,
        fontweight='bold',
        color='black'
    )

# Styling
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.box(False)

# Show plot
plt.show()
````

##### Chart 38 - Top 10 Stores By Sales

````python
# Chart - 38 Vizualisation Code

# Features Used - Store, Sales

# ✅ Why this chart is important to include?
# I am plotting the top 10 stores by total sales with store ID on the x-axis and sales on the y-axis.
# This layout is better for ranking (left-to-right) and is easier to read when store IDs are primary labels.

# ---------------------------------------------------------------------------------------------------------- #

# Preparing the top-10 data (summing sales per store)
top10 = (
    df.groupby('Store', as_index=False)['Sales']
      .sum()
      .sort_values('Sales', ascending=False)
      .head(10)
)

# Converting Store to string (safe for plotting) and keeping current ranking order
top10['Store'] = top10['Store'].astype(str)
top10 = top10.reset_index(drop=True)

# Scale sales to millions for nicer axis ticks and annotations
top10['Sales_Million'] = top10['Sales'] / 1_000_000

# Setting the visual style
sns.set(style='whitegrid')

# Create figure (vertical bars)
plt.figure(figsize=(12,6))
ax = sns.barplot(
    data=top10,
    x='Store',
    y='Sales_Million',
    palette='Blues_d'
)

# Title and axis labels
plt.title('Top 10 Stores by Total Sales', fontsize=16, fontweight='bold')
plt.xlabel('Store ID', fontsize=12)
plt.ylabel('Total Sales (Millions)', fontsize=12)

# Rotate x labels for readability
plt.xticks(rotation=45, ha='right')

# Annotate each bar with sales value (safe offset: 1.5% of max)
max_m = top10['Sales_Million'].max()
y_offset = max_m * 0.015  # 1.5% of max value

for i, row in top10.iterrows():
    ax.text(
        i,                                 # x-position (bar index)
        row['Sales_Million'] + y_offset,   # y-position (just above the bar)
        f"{row['Sales_Million']:.2f}M",    # text
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='semibold'
    )

# Tight layout and show
plt.tight_layout()
plt.show()
````

**Summary -**

* Store 262 leads with the highest total sales: ₹19.52M

* Next strong performers are 817 (₹17.06M), 562 (₹16.93M), and 1114 (₹16.20M)

* The top 10 stores all lie in the ₹12.9M – ₹19.5M range

* These stores consistently outperform the rest of the chain and are likely located in high-traffic or high-demand regions

* The gap between Rank 1 and Rank 10 is ~₹6.6M, indicating meaningful variation in store-level performance

##### Chart 41 - Top 10 Stores by Conversion Rate

````python
# Chart - 41 Visualisation Code

# Features Used - Store, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart is showing me which stores convert their customer footfall into actual revenue most effectively.
# A store might not have the highest number of customers, but still generate high revenue due to a strong conversion rate.
# Identifying such stores helps reveal operational excellence and premium customer behaviour.

# ---------------------------------------------------------------------------------------------------------- #

# Creating conversion rate metric: Total Sales / Total Customers per store
conversion_df = (
    df.groupby('Store', as_index=False)
      .agg({'Sales': 'sum', 'Customers': 'sum'})
)

# Avoiding division errors
conversion_df = conversion_df[conversion_df['Customers'] > 0]

# Calculating conversion rate
conversion_df['ConversionRate'] = conversion_df['Sales'] / conversion_df['Customers']

# Selecting top 10 stores by conversion rate
top10_conversion = (
    conversion_df.sort_values('ConversionRate', ascending=False)
                 .head(10)
)

# Converting store IDs to string for cleaner labeling
top10_conversion['Store'] = top10_conversion['Store'].astype(str)

# Setting plot style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(12,6))
ax = sns.barplot(
    data=top10_conversion,
    x='Store',
    y='ConversionRate',
    palette='Blues_d'
)

# Adding titles and labels
plt.title('Top 10 Stores by Conversion Rate (Sales per Customer)', fontsize=16, fontweight='bold')
plt.xlabel('Store ID', fontsize=12)
plt.ylabel('Conversion Rate (Sales ÷ Customers)', fontsize=12)

# Rotating x-axis labels for readability
plt.xticks(rotation=45, ha='right')

# Annotating values above each bar
max_val = top10_conversion['ConversionRate'].max()
y_offset = max_val * 0.03

for index, row in top10_conversion.iterrows():
    plt.text(
        index,
        row['ConversionRate'] + y_offset,
        f"{row['ConversionRate']:.2f}",
        ha='center',
        fontsize=10,
        fontweight='semibold'
    )

# Tight layout for cleaner appearance
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* The top 10 stores with the highest conversion rate (Sales ÷ Customers) fall in the range of 14.3 to 16.3, which means these stores generate exceptionally high revenue per customer visit.

* Stores 842, 612, 455, and 158 lead the list, achieving conversion rates above 15, indicating strong sales efficiency.

* Many of these stores do not appear in the top 10 sales or top 10 customer traffic charts, showing that despite having moderate footfall, they excel at converting customers into buyers.

* These stores represent high-efficiency units, likely serving premium customers, offering high-value assortments, or having more effective sales execution.

###### 1. Why did you pick the specific chart?

* A store might get fewer customers, but if each visitor buys more or buys higher-value items, the store becomes extremely profitable. This chart helps me detect such high-efficiency stores that were not obvious from the sales or customer charts alone.

🔹 2. These stores likely serve premium or high-spending customers

##### Chart 42 - Bottom 10 Stores by Conversion Rate

````python
# Chart - 42 Visualisation Code

# Features Used: Store, Sales, Customers

# -------------------------------------------------------------------------------------------------
# 📌 Why this chart is important to include?
# Conversion Rate = Total Sales ÷ Total Customers
# It helps us identify stores where footfall is NOT translating into revenue.
# These stores may have:
# - poor store layout
# - weak assortment
# - inefficient pricing or promotion strategy
# - customer-service issues
# - operational inefficiencies
# Finding these underperformers is critical for business decisions and corrective actions.
# -------------------------------------------------------------------------------------------------

# 🧮 Step 1: Aggregating Sales & Customers per Store
store_perf = df.groupby('Store').agg({
    'Sales': 'sum',
    'Customers': 'sum'
}).reset_index()

# Avoiding division by zero
store_perf = store_perf[store_perf['Customers'] > 0]

# 🧮 Step 2: Calculating Conversion Rate
store_perf['ConversionRate'] = store_perf['Sales'] / store_perf['Customers']

# 🧮 Step 3: Selecting Bottom 10 Stores
bottom_10 = store_perf.nsmallest(10, 'ConversionRate')

# -------------------------------------------------------------------------------------------------
# 🎨 Visualization - Vertical Bar Chart
# -------------------------------------------------------------------------------------------------

plt.figure(figsize=(12, 6))
sns.barplot(
    data=bottom_10,
    x='Store',
    y='ConversionRate',
    palette='OrRd_r'   # reverse orange-red scale for low → very low
)

# 📌 Adding value labels
for index, row in bottom_10.iterrows():
    plt.text(
        x=index,
        y=row['ConversionRate'] + 0.05,
        s=f"{row['ConversionRate']:.2f}",
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# Titles & Labels
plt.title("Bottom 10 Stores by Conversion Rate (Sales per Customer)", fontsize=16, fontweight='bold')
plt.xlabel("Store ID", fontsize=12)
plt.ylabel("Conversion Rate (Sales ÷ Customers)", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)


# Displaying the plot
plt.show()
````

##### Chart 43 - Percentage of Promo Days Across Days of the Week

````python
# Chart 43 - Vizualisation Code

# Features Used: DayOfWeek, Promo

# ✅ Why this chart is important to include?
# The earlier chart (DayOfWeek vs Promo count) showed how many promo records exist,
# but this percentage-based view helps me understand if promotions are used more
# aggressively on certain weekdays. This avoids bias caused by uneven record counts.

#----------------------------------------------------------------------------------------------------------#

# Preparing data for promo percentage per weekday
weekday_promo_rate = (
    df.groupby('DayOfWeek')['Promo'].mean().reset_index()
)

# Convert to percentage
weekday_promo_rate['PromoPercent'] = weekday_promo_rate['Promo'] * 100

# Setting style
sns.set(style="whitegrid")

# Creating the figure
plt.figure(figsize=(10,6))

# Barplot for promo % by weekday
sns.barplot(
    data=weekday_promo_rate,
    x='DayOfWeek',
    y='PromoPercent',
    palette='coolwarm'
)

# Adding labels above bars
for i, row in weekday_promo_rate.iterrows():
    plt.text(
        i,
        row['PromoPercent'] + 1,
        f"{row['PromoPercent']:.1f}%",
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# Titles and labels
plt.title("Percentage of Promo Days Across the Week", fontsize=16, fontweight='bold')
plt.xlabel("Day of the Week (1 = Monday, 7 = Sunday)", fontsize=12)
plt.ylabel("Promo Percentage (%)", fontsize=12)

# Display
plt.show()
````

* Weekends (Saturday and Sunday) have 0% promo days, confirming that promotions are not scheduled on weekends.

###### 1. Why did you pick the specific chart?

* It clearly highlights promo strategy patterns, especially weekday vs weekend discrepancies.

* Percentages help business teams compare promo consistency and identify whether certain weekdays are being prioritized more.

* This signals - Stores do not run promotions on weekends, likely due to:

✔ No weekday stands out as a “special promo day”

````python
# Chart - 45 Visualisation Code

# Features Used: StoreType, Promo

# ✅ Why this chart is important to include?
# This chart is showing how frequently each store type runs short-term promotions.
# Because raw counts can be misleading due to unequal store distribution,
# calculating the percentage allows us to understand each store type’s *strategic promotion intensity*.
# This helps in identifying which store formats rely heavily on promotions and which operate with fewer discounts.

# ------------------------------------------------------------------------------------------------------ #

# Preparing the data by calculating Promo percentage for each store type
promo_rate_storetype = (
    df.groupby('StoreType')['Promo']
      .mean()
      .reset_index()
      .rename(columns={'Promo': 'PromoPercent'})
)

# Converting to percentage format
promo_rate_storetype['PromoPercent'] *= 100

# Setting visualization style
sns.set(style="whitegrid")

# Creating the figure
plt.figure(figsize=(10,6))

# Creating barplot to visualize promo percentages
sns.barplot(
    data=promo_rate_storetype,
    x='StoreType',
    y='PromoPercent',
    palette='coolwarm'
)

# Adding data labels on bars
for idx, row in promo_rate_storetype.iterrows():
    plt.text(
        idx,
        row['PromoPercent'] + 1,
        f"{row['PromoPercent']:.1f}%",
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# Adding titles and labels
plt.title("Percentage of Promo Days Across Store Types", fontsize=16, fontweight='bold')
plt.xlabel("Store Type", fontsize=12)
plt.ylabel("Promo Percentage (%)", fontsize=12)

# Displaying the plot
plt.show()
````

##### Chart 47 - Percentage of Promo2 Active Days Across Each Month

````python
# Chart - 47 Visualisation

# Features Used: Month, IsPromo2Active

# ✅ Why this chart is important to include?
# This chart is showing the percentage of days in each month when Promo2 is active.
# Since Promo2 is a long-term recurring promotion, understanding its monthly activation
# helps identify how often stores participate in Promo2 and whether it follows
# a seasonal pattern. This is important for analyzing long-term promotional strategy
# and its possible effect on sales trends.


# ---------------------------------------------------------------------------------------------------------- #


# Calculating the percentage of Promo2 active days per month
promo2_percentage = (
    df.groupby("Month")["IsPromo2Active"]
    .mean() * 100
)

# Setting plot style
sns.set(style="whitegrid")

# Creating the figure
plt.figure(figsize=(12, 6))

# Plotting the bar chart for Promo2 activation percentage
sns.barplot(
    x=promo2_percentage.index,
    y=promo2_percentage.values,
    palette="Blues_r"
)

# Adding labels above each bar
for i, value in enumerate(promo2_percentage.values):
    plt.text(
        i,
        value + 1,        # Positioning text slightly above the bar
        f"{value:.1f}%",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

# Setting titles and axis labels
plt.title("Percentage of Promo2 Active Days Across Each Month", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Promo2 Active Percentage (%)", fontsize=12)

# Adjusting the y-limit for better text visibility
plt.ylim(0, promo2_percentage.max() + 10)

# Displaying the chart
plt.show()
````

**Summary -**

* Promo2 activation is highly uneven throughout the year.
The highest activation occurs in January (24.2%), April (24.2%), and July (23.3%), suggesting quarterly campaign cycles.

* Months like February, May, August, and November show moderate activation around 8-10%.

* The lowest activation appears in September (6.5%), December (6.5%), and June (7.3%), indicating intentional pauses, possibly to avoid campaign clutter or due to budget cycles.

* Overall, Promo2 follows a seasonal rhythm, likely aligned with strategic long-term promotions.


###### 1. Why did you pick the specific chart?

##### Chart 48 - Percentage of Promo2 Active Days Across Store Types

````python
# Chart - 48 Visualisation Code

# Features Used: StoreType, IsPromo2Active

# ✅ Why this chart is important to include?
# This chart is helping in understanding how long-term promotion (Promo2) is distributed
# across different store types. Unlike short-term promos, Promo2 is recurring and strategic.
# Checking its percentage usage across store types is helping us learn whether the company
# follows a uniform strategy or prefers certain store types for long-term promotional campaigns.

# -------------------------------------------------------------------------------------------------------- #

# Calculating Promo2 activation percentage for each store type
promo2_storetype_rate = (
    df.groupby('StoreType')['IsPromo2Active']
      .mean()
      .reset_index()
)

# Converting from rate to percentage
promo2_storetype_rate['Promo2Percent'] = promo2_storetype_rate['IsPromo2Active'] * 100

# Setting visual style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Plotting bar chart for Promo2 percentage across store types
sns.barplot(
    data=promo2_storetype_rate,
    x='StoreType',
    y='Promo2Percent',
    palette='Blues'
)

# Adding percentage labels on each bar
for idx, row in promo2_storetype_rate.iterrows():
    plt.text(
        idx,
        row['Promo2Percent'] + 0.5,
        f"{row['Promo2Percent']:.1f}%",
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# Titles and axis labels
plt.title("Percentage of Promo2 Active Days Across Store Types",pad=20, fontsize=16, fontweight='bold')
plt.xlabel("Store Type", fontsize=12)
plt.ylabel("Promo2 Active Percentage (%)", fontsize=12)

# Displaying the plot
plt.show()
````

* This indicates that Promo2 is not uniformly applied and is strategically favored for certain store formats.

##### Chart 49 - Percentage of Promo2 Active Days by Assortment Type

````python
# Chart - 49 Visualisation Code


# Features Used: Assortment, IsPromo2Active


# ✅ Why this chart is important to include?
# This chart is showing how frequently the long-term promotion (Promo2) is running
# for each assortment bucket. Promo2 is running over long windows — comparing
# its activation share across assortment types is revealing which product-mix
# segments are receiving sustained marketing support. This is helping identify
# whether Promo2 is being targeted equally across assortments or skewed to a few.

# --------------------------------------------------------------------------------------------------------- #

# Calculating percentage of Promo2 active days per assortment
promo2_by_assort = (
    df.groupby('Assortment')['IsPromo2Active']
      .mean()                    # mean is giving share (0-1) because column is 0/1
      .reset_index()
)
promo2_by_assort['Promo2Percent'] = promo2_by_assort['IsPromo2Active'] * 100

# Ensuring consistent assortment ordering (a, b, c) if present
order = ['a', 'b', 'c']
promo2_by_assort['Assortment'] = pd.Categorical(promo2_by_assort['Assortment'], categories=order, ordered=True)
promo2_by_assort = promo2_by_assort.sort_values('Assortment')

# Setting aesthetic style for plots
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(9,6))

# Drawing the barplot
sns.barplot(
    data=promo2_by_assort,
    x='Assortment',
    y='Promo2Percent',
    palette='Blues_d'
)

# Adding percentage labels above each bar
for idx, row in promo2_by_assort.iterrows():
    plt.text(
        x = list(promo2_by_assort['Assortment']).index(row['Assortment']),  # placing text at bar x-position
        y = row['Promo2Percent'] + 0.6,                                    # small offset above bar
        s = f"{row['Promo2Percent']:.1f}%",                                # formatted percentage
        ha = 'center',
        fontsize = 11,
        fontweight = 'bold'
    )

# Adding titles and axis labels
plt.title('Percentage of Promo2 Active Days Across Assortment Types', fontsize=16, fontweight='bold')
plt.xlabel('Assortment Type', fontsize=12)
plt.ylabel('Promo2 Active Percentage (%)', fontsize=12)

# Tight layout for cleaner rendering
plt.tight_layout()

# Displaying the plot
plt.show()
````

##### Chart 50 - Impact of Promotions on Daily Sales

````python
# Chart 51 - Visualisation Code

# Features Used: DayOfWeek, Promo, Sales

# ✅ Why this chart is important to include?
# This chart will show average sales for each day of the week separately for Promo and Non-Promo days.
# This will help in identifying which weekdays are benefiting most from promotions,
# and whether promo activity is delivering consistent uplift across the week or only on specific days.

# --------------------------------------------------------------- #


# Preparing the grouped data by DayOfWeek and Promo and calculating average sales
promo_day_sales = (
    df.groupby(['DayOfWeek', 'Promo'])['Sales']
      .mean()
      .reset_index()
      .sort_values(by='DayOfWeek')
)

# Creating a pivot table for easier line plotting (index = DayOfWeek, columns = Promo)
promo_day_pivot = promo_day_sales.pivot(
    index='DayOfWeek',
    columns='Promo',
    values='Sales'
)

# Renaming the columns for clarity (0 -> No Promo, 1 -> Promo)
promo_day_pivot.columns = ['No Promo', 'Promo']

# Ensuring day order is correct (1=Mon ... 7=Sun)
promo_day_pivot = promo_day_pivot.reindex([1,2,3,4,5,6,7])

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

# Plotting the line for Non-Promo average sales
plt.plot(
    promo_day_pivot.index,
    promo_day_pivot['No Promo'],
    marker='o',
    linewidth=2,
    label='No Promo',
    color='#1f77b4'
)

# Plotting the line for Promo average sales
plt.plot(
    promo_day_pivot.index,
    promo_day_pivot['Promo'],
    marker='o',
    linewidth=2,
    label='Promo',
    color='#ff7f0e'
)

# Filling the area between lines to visually emphasize the uplift (Promo - No Promo)
plt.fill_between(
    promo_day_pivot.index,
    promo_day_pivot['No Promo'],
    promo_day_pivot['Promo'],
    where=(promo_day_pivot['Promo'] >= promo_day_pivot['No Promo']),
    interpolate=True,
    color='#ffbb78',
    alpha=0.12
)

# Adding plot title, labels and legend
plt.title("Average Sales by Day of the Week: Promo vs Non-Promo", fontsize=16, fontweight='bold')
plt.xlabel("Day of the Week (1 = Monday, 7 = Sunday)", fontsize=12)
plt.ylabel("Average Sales", fontsize=12)
plt.legend(title="Promo (0 = No, 1 = Yes)")

# Adding grid and readable x-ticks (Monday...Sunday)
plt.grid(alpha=0.25)
plt.xticks(ticks=[1,2,3,4,5,6,7], labels=['Mon','Tue','Wed','Thu','Fri','Sat','Sun'])

# Applying tight layout for better spacing
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* The chart compares average sales on each weekday between promo and non-promo periods.

* The gap between the two lines clearly shows how promotions lift weekday sales.

* Promo sales remain consistently higher across Monday-Saturday, with Monday performing the strongest under promotions.

* Sunday, however, shows zero promo activity, which explains the extremely low non-promo sales point — stores are either closed or operate under restricted hours.

* Promotions appear most effective early in the week (Mon- Wed), and the uplift gradually declines toward Friday.

* Saturday also shows a healthy promo uplift, suggesting weekend shopping benefits from discounts.

* This chart helps evaluate when promotions produce the best return and highlights days where promo strategy might need rethinking.

###### 1. Why did you pick the specific chart?

* I selected this chart because it helps me understand how weekday performance changes depending on whether promotions are running or not.

* Since customer behaviour varies heavily by day of the week, this visualization will allow me to see which days respond strongly to promotions and which days are naturally weak or strong without promotions.

* This insight will guide targeted promo planning rather than using a uniform weekly approach.

###### 2. What is/are the insight(s) found from the chart?

* Promotions increase sales significantly on every weekday from Monday to Saturday.

* Monday shows the highest promo-driven boost, indicating customers respond strongly to discounting at the start of the week.

* The uplift gradually declines from Tue → Wed → Thu → Fri.

* Saturday also benefits from promo activity, showing solid uplift.

* Sunday has no promo activity, resulting in extremely low non-promo sales — likely due to store closures or partial operations.

* Since the promo uplift is not constant across days, the efficiency of running promos varies by weekday.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**✅ Positive Business Impact :**

* The insights will help determine which weekdays deserve stronger promotional spending (Monday, Tuesday, Saturday).

* Promo investment can be reduced on lower-responding days, improving ROI.

* Understanding weekday responsiveness can help scheduling staff, inventory, and operational activities more effectively.


**⚠️ Negative-Growth Risks Identified :**

* Relying on promotions heavily Monday-Friday may train customers to delay purchases until promo days.

* Non-promo weekday sales are significantly lower, indicating stores may be becoming promotion-dependent.

* Sunday underperformance might require either policy review or operational adjustment if stores are partially open.

##### Chart 51 - Impact of Promo on Daily Customer Footfall

````python
# Chart 51 - Visualisation Code

# Features Used: DayOfWeek, Promo, Customers

# ✅ Why this chart is important to include?
# This chart will be helping to compare average customer footfall on each weekday
# This chart will be showing whether promotions are shifting customer traffic to specific weekdays
# This chart will be guiding decisions about when to schedule long/short promos to maximise footfall

# --------------------------------------------------------------- #
# Preparing data: grouping by DayOfWeek and Promo to compute average customers
# Adding groupby that is calculating mean customers per DayOfWeek x Promo pair
daily_customers_promo = (
    df.groupby(['DayOfWeek', 'Promo'])['Customers']
      .mean()
      .reset_index()
)

# Creating a pivot to get a clean table for plotting (rows = day, cols = promo flag)
# Pivoting so that column 0 -> No Promo, 1 -> Promo
customers_pivot = daily_customers_promo.pivot(
    index='DayOfWeek',
    columns='Promo',
    values='Customers'
)

# Renaming pivot columns for readability
customers_pivot.columns = ['No Promo', 'Promo']

# Ensuring days are sorted 1..7 (Mon..Sun)
customers_pivot = customers_pivot.reindex([1,2,3,4,5,6,7])


# Setting plot style
sns.set(style="whitegrid")

# Creating figure
plt.figure(figsize=(12,6))

# Plotting No Promo line
plt.plot(
    customers_pivot.index,
    customers_pivot['No Promo'],
    marker='o',
    linewidth=2,
    label='No Promo',
    color='#1f77b4'
)

# Plotting Promo line
plt.plot(
    customers_pivot.index,
    customers_pivot['Promo'],
    marker='o',
    linewidth=2,
    label='Promo',
    color='#ff7f0e'
)

# Shading area between lines to emphasise uplift in customers during promo
plt.fill_between(customers_pivot.index,
                 customers_pivot['No Promo'],
                 customers_pivot['Promo'],
                 color='#ffbb78',
                 alpha=0.12)

# Adding titles and axis labels
plt.title("Average Customers by Day of Week: Promo vs Non-Promo", fontsize=16, fontweight='bold')
plt.xlabel("Day of the Week (1 = Monday, 7 = Sunday)", fontsize=12)
plt.ylabel("Average Number of Customers", fontsize=12)

# Custom x-ticks as weekday names
plt.xticks(ticks=[1,2,3,4,5,6,7], labels=['Mon','Tue','Wed','Thu','Fri','Sat','Sun'])

# Adding legend with descriptive title
plt.legend(title="Promo (0 = No, 1 = Yes)")

# Adding grid for readability and tightening layout
plt.grid(alpha=0.25)
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* The chart clearly shows that promotions consistently increase customer footfall across all weekdays except Sunday, where stores operate very differently and promo data is practically absent.

* Promo-driven uplift is strongest on Monday, where customers rise from ~670 to ~940 on promotional days.

* On Thursday and Friday, the gap between promo and non-promo customers narrows, indicating reduced promo sensitivity toward the weekend.

* Saturday and Sunday behave uniquely—Saturday has moderate decline from Friday, while Sunday collapses due to very low/no store operation or missing values.

* Overall, the promo uplift pattern is weekday-dependent, showing clear opportunities to strategically place promotions for maximum footfall.

###### 1. Why did you pick the specific chart?

* I selected a line chart because it is the most intuitive way to show changes and trends across ordered categories, such as days of the week.

* I chose the features DayOfWeek, Promo, and Customers because this relationship is crucial for understanding when promotions work best—not just whether they work.

* This chart allows me to visually inspect behavioural patterns in customer footfall and identify weekday-specific promo responsiveness.

##### Chart 53 - Monthly Sales Comparison Promo vs Non-Promo

````python
# Chart - 53 Visualization Code

# Features Used: Month, Sales, Promo

# ✅ Why this chart is important to include?
# This chart is helping in comparing average monthly sales under Promo days versus Non-Promo days.
# By separating sales trends into two lines (Promo = 1 and Promo = 0), this visualization is showing
# how promotions influence monthly sales patterns. It is helping in identifying months where promotions
# bring strong uplift and months where promo impact is weak, supporting seasonal marketing decisions.

# -------------------------------------------------------------------------------------------------------- #

# Creating month name mapping for cleaner x-axis labels
month_map = {1:'Jan',2:'Feb',3:'Mar',4:'Apr',5:'May',6:'Jun',
             7:'Jul',8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}

# Adding a MonthName column for better readability
df['MonthName'] = df['Month'].map(month_map)

# Preparing aggregated data: calculating average monthly sales separately for Promo = 0 and Promo = 1
monthly_promo_sales = (
    df.groupby(['Month', 'MonthName', 'Promo'])['Sales']
      .mean()
      .reset_index()
      .rename(columns={'Sales': 'AvgSales'})
)

# Ordering month names correctly on x-axis
ordered_months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
monthly_promo_sales['MonthName'] = pd.Categorical(monthly_promo_sales['MonthName'],
                                                  categories=ordered_months,
                                                  ordered=True)

# Setting visualization style
sns.set(style="whitegrid")

# Creating the figure
plt.figure(figsize=(12,6))

# Plotting the lineplot showing Promo vs Non-Promo monthly sales
sns.lineplot(
    data=monthly_promo_sales,
    x='MonthName',
    y='AvgSales',
    hue='Promo',
    marker='o',
    linewidth=2.5,
    palette=['#2E86C1', '#F39C12']      # Blue for No Promo, Orange for Promo
)

# Adding title and axis labels
plt.title("Average Monthly Sales: Promo vs Non-Promo", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Average Sales", fontsize=12)

# Customizing legend
plt.legend(title="Promo (0 = No, 1 = Yes)")

# Displaying the plot
plt.tight_layout()
plt.show()
````

**Summary -**

* The chart clearly shows that sales are consistently higher on Promo days compared to Non-Promo days across all months.

* The uplift is especially strong in June, July, November, and dramatically in December, where Promo sales peak sharply.

* Non-Promo sales remain stable throughout the year, with minor seasonal bumps, but they never come close to Promo-day performance.

* This tells us that promotions significantly boost revenue and the business relies heavily on promotion-based traffic and conversions.

##### Chart 54 - Impact of Promo on Monthly Customer Footfall

````python
# Chart 54 - Visualisation Code

# Features Used: Month, Promo, Customers

# ✅ Why this chart is important to include?
# This chart will help in comparing monthly customer footfall between Promo and Non-Promo periods.
# It will reveal whether promotions are driving additional store visits or if customer traffic follows seasonal patterns.
# This will help business teams understand whether promotions are improving customer acquisition or merely boosting sales from existing visitors.

# --------------------------------------------------------------- #
# Grouping the data by Month and Promo and calculating average customers
monthly_customer_promo = (
    df.groupby(['Month', 'Promo'])['Customers']
      .mean()
      .reset_index()
)

# Creating a pivot table for easier line plotting
customer_pivot = monthly_customer_promo.pivot(
    index='Month',
    columns='Promo',
    values='Customers'
)

# Renaming the columns for clarity
customer_pivot.columns = ['No Promo', 'Promo']

# Sorting the index to ensure months appear in correct order
customer_pivot = customer_pivot.sort_index()

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

# Plotting line for Non-Promo customer averages
plt.plot(
    customer_pivot.index,
    customer_pivot['No Promo'],
    marker='o',
    linewidth=2,
    label='No Promo',
    color='#1f77b4'
)

# Plotting line for Promo customer averages
plt.plot(
    customer_pivot.index,
    customer_pivot['Promo'],
    marker='o',
    linewidth=2,
    label='Promo',
    color='#ff7f0e'
)

# Adding title and labels
plt.title("Average Monthly Customers: Promo vs Non-Promo", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Average Customers", fontsize=12)

# Adding legend for clarity
plt.legend(title="Promo (0 = No, 1 = Yes)")

# Adding grid for readability
plt.grid(alpha=0.3)

# Displaying the plot
plt.show()
````

**Summary -**

* The uplift is very stable, indicating promotions reliably attract additional store visits.

* A clear seasonal spike appears in December, where Promo-day customer traffic jumps close to 1,000 customers, the highest among all months.

* Even Non-Promo traffic increases slightly in December, implying natural seasonal demand — but Promo magnifies this effect significantly.

* May shows the lowest Promo-day footfall (~760), suggesting weaker promotional response in early summer.


* Overall, Promo days clearly drive higher customer acquisition, and the effect strengthens during festive or peak-shopping months.

* I included this chart to understand how customer footfall changes across months when promotions are active vs inactive.

* Without this comparison, we wouldn't know if higher sales during promos are due to increased customers or simply increased spend per customer.


###### 2. What is/are the insight(s) found from the chart?

##### Chart 56 - Average Sales by Assortment Type: Promo vs Non-Promo

````python
# Chart 56 - Visualization Code

# Features Used: Promo, Assortment, Sales

# ✅ Why this chart is important to include?
# This chart helps analyze whether short-term daily promotions (Promo)
# perform differently across various product assortment types.
# Since assortment defines product variety and depth, promotions may
# not have a uniform impact on sales across assortments.
# A grouped bar chart enables a clear comparison of Promo vs No Promo
# within each assortment category.

# --------------------------------------------------------------- #

# Grouping data by Assortment and Promo to calculate average sales
assortment_promo_sales = (
    df.groupby(['Assortment', 'Promo'])['Sales']
      .mean()
      .reset_index()
)

# Pivoting data for grouped bar plotting
sales_pivot = assortment_promo_sales.pivot(
    index='Assortment',
    columns='Promo',
    values='Sales'
)

# Renaming columns for clarity
sales_pivot.columns = ['No Promo', 'Promo']

# --------------------------------------------------------------- #
# Visualization

plt.figure(figsize=(10, 8))

sales_pivot.plot(
    kind='bar',
    width=0.7,
    color=['#4C72B0', '#DD8452'],
    edgecolor='black'
)

# Adding value labels on bars
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f', padding=2, fontsize=9)

# Titles and labels
plt.title(
    'Average Sales by Assortment Type: Promo vs Non-Promo',
    fontsize=15,
    fontweight='bold'
)
plt.xlabel('Assortment Type', fontsize=12)
plt.ylabel('Average Sales', fontsize=12)

# Adjusting the alignment of tick labels
plt.xticks(rotation=0, ha='center')

# Legend formatting
plt.legend(title='Promo Status', frameon=True, loc = 'upper center',ncols =2, bbox_to_anchor=(0.5, -0.2))

# Grid and layout
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.box(False)
plt.tight_layout()

# Display plot
plt.show()
````

**Summary -**

* This chart shows that daily promotions consistently increase average sales across all assortment types, but the magnitude of uplift varies significantly by assortment.

* Assortment a and c experience a strong sales jump during Promo days, indicating that daily promotions are particularly effective when product variety is either limited (a) or well-balanced (c).

* Assortment b already performs well even without promotions, and Promo leads to only a marginal incremental gain, suggesting diminishing returns for heavy assortments.

* Overall, the chart highlights that daily Promo effectiveness is not uniform and is strongly influenced by the underlying assortment strategy.

###### 1. Why did you pick the specific chart?

* I chose a grouped bar chart because it allows a direct and intuitive comparison between Promo and Non-Promo sales within each assortment category.

* This chart helps explore whether daily promotional strategies interact differently with assortment depth, rather than assuming a one-size-fits-all Promo impact.

* By placing Promo and Non-Promo bars side by side, the chart clearly surfaces incremental sales lift patterns that would be harder to notice in aggregate or trend-based plots.

###### 2. What is/are the insight(s) found from the chart?

* Assortment a shows the highest relative sales uplift during Promo, indicating that limited assortments benefit strongly from short-term discounts.

* Assortment c also exhibits a substantial Promo-driven increase, suggesting that balanced assortments convert promotions into higher ticket values effectively.

* Assortment b shows the smallest gap between Promo and Non-Promo sales, implying that its strong baseline performance reduces Promo dependency.

###### 3. Will the gained insights help creating a positive business impact?

* The selected features — Promo, Assortment, and Customers — help explore whether promotions are equally effective across different product variety strategies.

###### 2. What is/are the insight(s) found from the chart?

##### Chart 64 - Sales Distribution by School Holiday: Promo vs Non-Promo

````python
# Chart - 64 Visualisation Code

# Features Used - Promo, SchoolHoliday, Sales

# ✅ Why this chart is important to include?
# This chart helps analyze how daily promotions behave during school holidays
# compared to non-holiday periods.
# Instead of relying only on average sales, a box plot captures the full
# sales distribution, including median behavior, variability, and spread.
# This is critical because school holidays often introduce high volatility
# in customer behavior, and promotions may cause inconsistent sales spikes
# rather than stable growth.

# ---------------------------------------------------------------------------------------------------------- #

# Setting consistent visual style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10, 6))

# Plotting box plot to analyze sales distribution
sns.boxplot(
    data=df,
    x='SchoolHoliday',
    y='Sales',
    hue='Promo',
    palette=['#64B5F6', '#FFB74D'],
    showfliers=False  # Removing extreme outliers for better readability
)

# Adding tick labels
plt.xticks(ticks = [0,1], labels = ["No Schhol Holiday", "School Holiday"])

# Annotate median values on box plot
grouped_medians = (
    df
    .groupby(['SchoolHoliday', 'Promo'])['Sales']
    .median()
    .reset_index()
)

for i, row in grouped_medians.iterrows():
    x_pos = row['SchoolHoliday']-0.2  if row['Promo'] == 0 else row['SchoolHoliday'] +0.2
    plt.text(
        x_pos,
        row['Sales'],
        f"{int(row['Sales'])}",
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold',
        color='black'
    )

# Adding chart title and labels
plt.title(
    'Sales Distribution by School Holiday: Promo vs Non-Promo',
    fontsize=15,
    fontweight='bold'
)
plt.xlabel('School Holibday', fontsize=12)
plt.ylabel('Sales', fontsize=12)

# Adjusting legend placement to avoid overlap
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(
    handles,
    ['No Promo', 'Promo Active'],
    title='Promo Status',
    loc='upper right',
    frameon=True
)


# Adding gridlines and finishing touches
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.box(False)

# Displaying the chart
plt.show()
````

**Summary -**

* This box plot shows that Promo days consistently outperform Non-Promo days in terms of median sales, regardless of whether it is a school holiday or not.

* During school holidays, both Promo and Non-Promo sales medians increase, indicating higher baseline demand driven by holiday-driven footfall.

* The highest median sales (~7.5K+) occur when Promo is active, with slightly higher medians during school holidays, suggesting a compounding effect of Promo + Holiday.

* Promo days also show wider interquartile ranges and longer upper whiskers, highlighting higher sales variability and occasional demand spikes.

###### 1. Why did you pick the specific chart?

* This box plot was chosen because it allows comparison of entire sales distributions, not just averages.

* It clearly highlights median shifts, variability, and extreme values, which are critical for understanding promotion performance during holidays.

* Unlike bar charts, this visualization helps identify risk and volatility, which is important for operational and inventory planning.

###### 2. What is/are the insight(s) found from the chart?

* Promo increases median sales significantly in both cases:

    * No School Holiday: Median rises from ~4.5K (No Promo) to ~7.5K (Promo).

    * School Holiday: Median rises from ~5.2K (No Promo) to ~7.6K (Promo).

* School holidays independently increase sales even without Promo, but Promo amplifies this effect.

* Promo days have higher dispersion, indicating larger upside potential but less predictability.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

* **No direct negative growth** is observed in terms of median sales; Promo consistently improves central sales tendency.

* However, higher variability during Promo days suggests potential operational risk—poor stock planning or staffing could lead to lost sales opportunities or inefficiencies.

* Non-Promo days during non-holiday periods show lower medians and tighter ranges, which may indicate stable but limited revenue potential.


##### Chart 65 - Impact of Promo and Competition Age on Sales

````python
# Chart - 65 Visualisation Code

# Features Used - Promo, CompetitionOpenMonths, Sales

# ✅ Why this chart is important to include?
# This chart helps analyze how the effectiveness of daily promotions (Promo)
# varies depending on how long a competing store has been operating nearby.
# It allows us to understand whether promotions are more effective when
# competition is new, moderate, or well-established.
# ---------------------------------------------------------------------------------------------------------- #

# Creating competition age buckets (in months)
df['Competition_Age_Bucket'] = pd.cut(
    df['CompetitionOpenMonths'],
    bins=[-1, 6, 12, 24, 60, 120, np.inf],
    labels=[
        'Very New (<6m)',
        'New (6–12m)',
        'Moderate (1–2y)',
        'Old (2–5y)',
        'Very Old (5–10y)',
        'Extremely Old (>10y)'
    ]
)

# Aggregating average sales by competition age bucket and promo status
promo_comp_sales = (
    df.groupby(['Competition_Age_Bucket', 'Promo'])['Sales']
      .mean()
      .reset_index()
)

# Setting plot style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(12, 6))

# Plotting grouped bar chart
sns.barplot(
    data=promo_comp_sales,
    x='Competition_Age_Bucket',
    y='Sales',
    hue='Promo',
    palette=['#4E79A7', '#F28E2B']
)

# Adding value labels on bars
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f', fontsize=12)

# Titles and labels
plt.title('Average Sales by Competition Age: Promo vs Non-Promo', fontsize=14)
plt.xlabel('Competition Age (Months Since Competitor Opened)', fontsize=11)
plt.ylabel('Average Sales', fontsize=11)

# Legend formatting
plt.legend(handles= ax.containers,
    title='Promo Status',
    labels=['No Promo', 'Promo'],
    loc='upper center',
    bbox_to_anchor = [0.5,-0.1],
    ncols = 2
)

# Rotating x-axis labels for readability
plt.xticks(rotation=0)

# Tight layout for clean rendering
plt.tight_layout()

# Display plot
plt.show()
````

**Summary -**

* Across all competition maturity levels, promotional days consistently generate significantly higher sales than non-promotional days.

* While base sales (No Promo) slightly decline as competition ages, Promo maintains a strong and stable uplift, indicating that daily promotions remain effective even in highly competitive environments.

##### Chart 67 - Average Monthly Sales: Promo2 Active vs Not Active

````python
# Chart - 67 Visualisation Code

# Features Used: Month, Promo2, Sales

# ✅ Why this chart is important to include?
# This chart is going to help in understanding how long-running promotions (Promo2)
# influence monthly sales trends. Unlike short promotional spikes, Promo2 is meant
# to deliver sustained improvement in revenue. By comparing Promo2-active vs
# non-active months, we are expecting to uncover whether this long-term strategy
# truly contributes to higher monthly performance or follows seasonal patterns.

# --------------------------------------------------------------- #
# Grouping the data by Month and Promo2 and calculating average sales
monthly_sales_promo2 = (
    df.groupby(['Month', 'Promo2'])['Sales']
      .mean()
      .reset_index()
)

# Creating a pivot table for line plotting
sales_pivot_promo2 = monthly_sales_promo2.pivot(
    index='Month',
    columns='Promo2',
    values='Sales'
)

# Renaming columns for clarity
sales_pivot_promo2.columns = ['No Promo2', 'Promo2 Active']

# Sorting months in chronological order
sales_pivot_promo2 = sales_pivot_promo2.sort_index()

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

# Plotting line for months without Promo2
plt.plot(
    sales_pivot_promo2.index,
    sales_pivot_promo2['No Promo2'],
    marker='o',
    linewidth=2,
    label='No Promo2',
    color='#1f77b4'
)

# Plotting line for months with Promo2 active
plt.plot(
    sales_pivot_promo2.index,
    sales_pivot_promo2['Promo2 Active'],
    marker='o',
    linewidth=2,
    label='Promo2 Active',
    color='#ff7f0e'
)

# Adding shaded area to visualize the gap
plt.fill_between(
    sales_pivot_promo2.index,
    sales_pivot_promo2['No Promo2'],
    sales_pivot_promo2['Promo2 Active'],
    color='#ffbb78',
    alpha=0.15
)

# Adding titles, labels, and legend
plt.title("Average Monthly Sales: Promo2 Active vs Not Active", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Average Sales", fontsize=12)
plt.legend(title="Promo2 (0 = No, 1 = Yes)")

# Adding grid and month labels
plt.grid(alpha=0.25)
plt.xticks(
    ticks=sales_pivot_promo2.index,
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)


# Displaying the plot
plt.tight_layout()
plt.show()
````

**Summary -**

* Sales remain consistently higher during months without Promo2, suggesting that long-running promotions may not be delivering the intended uplift.

* The average sales gap between No Promo2 vs Promo2 Active is visible in every month, showing a structurally weaker performance under long-term promotions.

* The seasonal trend peaks in July and December, and even during these naturally high-demand periods, Promo2 months still underperform compared to Non-Promo2 months.

* Promo2 seems to behave more like a maintenance strategy rather than a revenue-boosting strategy—customers may not be responding strongly to long-running offers.

* This indicates that Promo2 may require better targeting, improved offer structure, or stronger marketing communication to be effective.

###### 1. Why did you pick the specific chart?

* I selected this chart to evaluate whether long-running promotions (Promo2) actually improve sales performance when viewed across the monthly business cycle.

* I chose a dual-line time series plot because it clearly separates two behavioral groups (Promo2 Active vs Not Active) and allows us to compare seasonal sales trajectories side by side.

* A line plot is the best choice here because Promo2 runs on monthly cycles, and line charts visually capture trends, fluctuations, and gaps over time that would be missed in bar or box plots.

* Using Month, Promo2, and Sales together helps uncover whether Promo2 aligns with natural demand cycles or works independently of seasonality.


###### 2. What is/are the insight(s) found from the chart?

* Sales are consistently higher in all months when Promo2 is NOT active, showing that long-term promotions are not generating the expected uplift.

##### Chart 68 - Promo2 Impact on Customer Footfall Across Months

````python
# Chart 68 - Visualization Code

# Features Used: Month, Promo2, Customers

# --------------------------------------------------------------- #
# Grouping data by Month and Promo2 and calculating total customers per group
monthly_customers = (
    df.groupby(['Month', 'Promo2'])['Customers']
      .sum()
      .reset_index()
)

# Calculating total customers per month
monthly_total = (
    df.groupby('Month')['Customers']
      .sum()
      .reset_index()
      .rename(columns={'Customers': 'TotalCustomers'})
)

# Merging totals back
monthly_customers = monthly_customers.merge(monthly_total, on='Month', how='left')

# Computing percentages
monthly_customers['CustomerPercent'] = (
    monthly_customers['Customers'] / monthly_customers['TotalCustomers'] * 100
)

# Pivoting for grouped bar chart
promo2_percent_pivot = monthly_customers.pivot(
    index='Month',
    columns='Promo2',
    values='CustomerPercent'
)

# Ensuring months 1–12 appear
all_months = list(range(1,13))
promo2_percent_pivot = promo2_percent_pivot.reindex(all_months).fillna(0)

# Renaming columns for readability
cols = {}
if 0 in promo2_percent_pivot.columns:
    cols[0] = 'No Promo2'
if 1 in promo2_percent_pivot.columns:
    cols[1] = 'Promo2 Active'

promo2_percent_pivot = promo2_percent_pivot.rename(columns=cols)

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

x = np.arange(len(promo2_percent_pivot.index))
width = 0.36

# Bars for No Promo2
if 'No Promo2' in promo2_percent_pivot.columns:
    plt.bar(
        x - width/2,
        promo2_percent_pivot['No Promo2'],
        width,
        label='No Promo2',
        color='#1f77b4'
    )

# Bars for Promo2 Active
if 'Promo2 Active' in promo2_percent_pivot.columns:
    plt.bar(
        x + width/2,
        promo2_percent_pivot['Promo2 Active'],
        width,
        label='Promo2 Active',
        color='#ff7f0e'
    )

# Adding labels above each bar
for idx in range(len(x)):
    for col, shift in [('No Promo2', -width/2), ('Promo2 Active', width/2)]:
        if col in promo2_percent_pivot.columns:
            val = promo2_percent_pivot[col].iloc[idx]
            if val > 0:
                plt.text(
                    x[idx] + shift,
                    val + 0.6,
                    f"{val:.1f}%",
                    ha='center',
                    fontsize=9
                )

# Titles, labels, and ticks
plt.title("Percentage of Monthly Customer Footfall: Promo2 Active vs Non-Active", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Customer Percentage (%)", fontsize=12)
plt.xticks(
    ticks=x,
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)


# Adding Legend
plt.legend(
    title="Promo2 Status",
    loc='upper center',
    bbox_to_anchor=(0.5, -0.12),   # Moves legend below chart
    ncol=2,                        # Two columns (side-by-side)
    frameon=False,                 # Removes legend border
    fontsize=10,
    title_fontsize=11
)

# Grid
plt.grid(axis='y', alpha=0.25)
plt.tight_layout()

plt.show()
````

**Summary -**

* The chart clearly shows that across all months, a higher share of customer footfall comes from non-Promo2 days compared to Promo2 active days.

* Promo2 contributes consistently lower customer percentages, typically ranging between 42%–46%, while non-Promo2 days contribute around 54%–57%.

* The gap between Promo2 and non-Promo2 customer share remains stable throughout the year, indicating that Promo2 does not significantly alter customer visit behavior on a monthly basis

###### 1. Why did you pick the specific chart?

* I chose a grouped bar chart because it allows direct side-by-side comparison of customer contribution between Promo2 active and non-active days for each month.

* I selected percentage of customers instead of absolute counts to avoid misleading conclusions caused by differences in total monthly traffic.

* I included Month and Promo2 together because Promo2 is a long-running promotion, and this chart helps evaluate whether its presence actually shifts customer behavior over time.

* This chart is specifically helping to assess whether Promo2 drives incremental footfall or merely exists alongside normal demand.

###### 2. What is/are the insight(s) found from the chart?

Key Insight:

* Promo2 does not dominate customer footfall in any month.

* Across all months, non-Promo2 days consistently attract more customers than Promo2 active days.

* The relatively flat Promo2 percentage trend suggests that customers do not significantly change their visit patterns due to Promo2 being active.

* Even in traditionally high-traffic months like November and December, Promo2 does not close the gap with non-Promo2 footfall.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**✅ Positive Business Impact :**

* This insight helps the business set realistic expectations from long-term promotions.

* It suggests Promo2 may be more effective for retention or basket value rather than for attracting new customers.

**⚠️ Negative-Growth Risks Identified -**

##### Chart 69 - Monthly Sales Per Customer: Promo2 Active vs Non-Active

````python
# Chart 69 - Visualisation Code

# Features Used: Month, Promo2, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart is going to analyze how long-running promotions (Promo2) impact
# sales efficiency over time.
# Instead of looking at total sales, this chart is focusing on Sales Per Customer,
# which is going to reveal whether Promo2 is improving customer spending behavior
# or merely increasing footfall without meaningful revenue contribution.
# The month-wise trend is going to help detect seasonal patterns and long-term
# effectiveness of Promo2.

# --------------------------------------------------------------- #


# Grouping data by Month and Promo2 and calculating average Sales Per Customer
monthly_spc_promo2 = (
    df.groupby(['Month', 'Promo2'])['SalesPerCustomer']
      .mean()
      .reset_index()
)

# Creating pivot table for line plot
spc_pivot = monthly_spc_promo2.pivot(
    index='Month',
    columns='Promo2',
    values='SalesPerCustomer'
)

# Renaming columns for clarity
spc_pivot.columns = ['No Promo2', 'Promo2 Active']

# Sorting months to ensure correct order
spc_pivot = spc_pivot.sort_index()

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

# Plotting Sales Per Customer for No Promo2
plt.plot(
    spc_pivot.index,
    spc_pivot['No Promo2'],
    marker='o',
    linewidth=2,
    label='No Promo2',
    color='#4C72B0'
)

# Plotting Sales Per Customer for Promo2 Active
plt.plot(
    spc_pivot.index,
    spc_pivot['Promo2 Active'],
    marker='o',
    linewidth=2,
    label='Promo2 Active',
    color='#DD8452'
)

# Adding filled area to emphasize gap
plt.fill_between(
    spc_pivot.index,
    spc_pivot['No Promo2'],
    spc_pivot['Promo2 Active'],
    alpha=0.12,
    color='#DD8452'
)

# Adding title and labels
plt.title(
    'Monthly Sales Per Customer: Promo2 Active vs Non-Active',
    fontsize=16,
    fontweight='bold'
)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales Per Customer', fontsize=12)

# Adding legend with improved placement
plt.legend(
    title='Promo2 Status',
    loc='upper left',
    frameon=True
)

# Adding grid and layout adjustments
plt.grid(alpha=0.25)
plt.xticks(
    ticks=spc_pivot.index,
    labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)
plt.tight_layout()

# Displaying the plot
plt.show()
````

##### Chart 70 -  Sales Change by Assortment Type: Promo2 vs No Promo2

````python
# Chart 70 - Visualisation Code

# Features Used: Assortment, Promo, Sales

# ✅ Why this chart is important to include?
# This chart is being included to compare how average sales change when promotions
# are applied across different assortment types.
# Using a slope chart is helping to clearly visualize the *direction and magnitude*
# of sales movement from No Promo to Promo for each assortment.
# This is allowing us to understand whether promotions are equally effective
# across assortment strategies or if certain assortments benefit more than others.

# --------------------------------------------------------------- #
# Grouping the data by Assortment and Promo and calculating average sales
assortment_sales_promo = (
    df.groupby(['Assortment', 'Promo'])['Sales']
      .mean()
      .reset_index()
)

# Creating a pivot table for slope chart structure
assortment_sales_pivot = assortment_sales_promo.pivot(
    index='Assortment',
    columns='Promo',
    values='Sales'
)

# Renaming columns for clarity
assortment_sales_pivot.columns = ['No Promo', 'Promo']

# Sorting assortment types for consistent ordering
assortment_sales_pivot = assortment_sales_pivot.sort_index()

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(11,6))

# Defining x-axis positions for slope lines
x_positions = [0.5, 1]

# Plotting slope lines for each assortment type
for assortment in assortment_sales_pivot.index:
    plt.plot(
        x_positions,
        assortment_sales_pivot.loc[assortment],
        marker='o',
        linewidth=2,
        label=f"Assortment {assortment}"
    )

    # Adding value labels for No Promo
    plt.text(
        x_positions[0] - 0.01,
        assortment_sales_pivot.loc[assortment, 'No Promo'],
        f"{assortment_sales_pivot.loc[assortment, 'No Promo']:.0f}",
        ha='right',
        va='center',
        fontsize=10
    )

    # Adding value labels for Promo
    plt.text(
        x_positions[1] + 0.02,
        assortment_sales_pivot.loc[assortment, 'Promo'],
        f"{assortment_sales_pivot.loc[assortment, 'Promo']:.0f}",
        ha='left',
        va='center',
        fontsize=10
    )

# Adding vertical reference lines
plt.axvline(x=x_positions[0], color='gray', linestyle='--', alpha=0.4)
plt.axvline(x=x_positions[1], color='gray', linestyle='--', alpha=0.4)

# Customizing x-axis labels
plt.xticks(x_positions, ['No Promo', 'Promo'])

# Adding titles and labels
plt.title(
    "Average Sales Change by Assortment Type: Promo vs No Promo",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Promotion Status", fontsize=12)
plt.ylabel("Average Sales", fontsize=12)

# Adding legend
plt.legend(title="Assortment Type")

# Adding grid for readability
plt.grid(axis='y', alpha=0.3)

# Adjusting layout
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* The chart clearly shows that promotions lead to an increase in average sales across all assortment types, but the magnitude of uplift varies significantly.

* Assortment A and C experience a strong jump in sales when promotions are applied, indicating high promo responsiveness.

* Assortment B already performs strongly even without promotions and shows comparatively limited incremental benefit from promos.

* This indicates that promotions are not uniformly effective and should be aligned with assortment strategy.

###### 1. Why did you pick the specific chart?

* I chose a slope chart because it is best suited for comparing changes between two states — No Promo and Promo — across multiple categories.

* This chart helps visually emphasize both the direction and magnitude of sales change rather than just absolute values.

* Using average sales instead of percentages preserves the real baseline performance and avoids misleading zero-reference distortions.

* The combination of Assortment + Promo + Sales allows me to understand whether assortment strategy influences promotional effectiveness.

###### 2. What is/are the insight(s) found from the chart?

* Assortment A shows a strong uplift from ~4,100 to ~7,700 average sales, indicating high sensitivity to promotions.

* Assortment C also demonstrates a substantial gain, moving from ~4,700 to ~8,300, suggesting promotions significantly enhance its revenue potential.

* Assortment B has the highest baseline sales (~8,300) but gains only a modest increase during promotions (~8,900).

* This suggests that Assortment B already attracts high-paying customers and is less dependent on discounts to drive sales.

Are there any insights that lead to negative growth? Justify with specific reason.

**✅ Positive Business Impact :**

##### Chart 73 - Impact of Promo2 on Sales by Store Type (% Change)

````python
# Chart 73 - Visualisation Code

# Features Used: StoreType, Promo2, Sales

# ✅ Why this chart is important to include?
# This chart is helping to understand how long-running promotions (Promo2)
# are impacting sales performance differently across store types.
# By converting absolute sales into percentage change, this chart is allowing
# fair comparison between store formats with very different baseline sales.
# This is helping to identify which store types are benefiting from Promo2
# and which store types are losing revenue due to prolonged discounting.

# --------------------------------------------------------------- #
# Grouping the data by StoreType and Promo2 and calculating average sales
storetype_promo2_sales = (
    df.groupby(['StoreType', 'Promo2'])['Sales']
      .mean()
      .reset_index()
)

# Creating a pivot table to separate Promo2 Active vs Non-Active
sales_pivot = storetype_promo2_sales.pivot(
    index='StoreType',
    columns='Promo2',
    values='Sales'
)

# Renaming columns for clarity
sales_pivot.columns = ['No Promo2', 'Promo2 Active']

# --------------------------------------------------------------- #
# Calculating percentage change due to Promo2
sales_pivot['Sales_Percent_Change'] = (
    (sales_pivot['Promo2 Active'] - sales_pivot['No Promo2'])
    / sales_pivot['No Promo2']
) * 100

# Resetting index for plotting
sales_pct_df = sales_pivot.reset_index()

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(10,6))

# Creating grouped bar chart for percentage change
sns.barplot(
    data=sales_pct_df,
    x='StoreType',
    y='Sales_Percent_Change',
    palette='coolwarm'
)

# Adding value labels on bars
for idx, row in sales_pct_df.iterrows():
    plt.text(
        idx,
        row['Sales_Percent_Change'] - 1,
        f"{row['Sales_Percent_Change']:.1f}%",
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# Adding reference line at 0%
plt.axhline(0, color='black', linewidth=1, linestyle='--')

# Adding titles and labels
plt.title(
    "Impact of Promo2 on Sales by Store Type (% Change)",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Store Type", fontsize=12)
plt.ylabel("Sales Percentage Change (%)", fontsize=12)

# Adding grid for readability
plt.grid(axis='y', alpha=0.3)

# Displaying the plot
plt.tight_layout()
plt.show()
````

**Summary -**

* The chart shows that Promo2 (long-running promotion) leads to a decline in average sales across all store types.

* Store type B is the most negatively impacted, with sales dropping by nearly 27.3% when Promo2 is active.

##### Chart 75 - Average Sales per Customer by Store Type: Promo2 Active vs Non-Active

##### Chart 77 - Sales Distribution (%) by Competition Age: Promo2 Active vs Non-Active

````python
# Chart 77 - Visualisation Code

# Features Used: CompetitionOpenMonths, Promo2, Sales

# ✅ Why this chart is important to include?
# This chart will help me compare how sales change during long-running promotions (Promo2)
# across different stages of competitive pressure.
# By representing sales impact in percentage terms, this chart will allow me
# to understand whether Promo2 is relatively improving or worsening sales
# when competition is new versus when competition has existed for a longer duration.
# Using a grouped bar chart will make the comparison between Promo2 and Non-Promo2
# clear and easy to interpret across competition age buckets.

# --------------------------------------------------------------- #
# Creating competition age buckets using months since competition opened
df['CompetitionAgeBucket'] = pd.cut(
    df['CompetitionOpenMonths'],
    bins=[-1, 6, 12, 24, 60, 120, df['CompetitionOpenMonths'].max()],
    labels=[
        'Very New (<6m)',
        'New (6–12m)',
        'Moderate (1–2y)',
        'Old (2–5y)',
        'Very Old (5–10y)',
        'Extremely Old (>10y)'
    ]
)

# Calculating average sales separately for Promo2 and Non-Promo2
promo2_sales = (
    df.groupby(['CompetitionAgeBucket', 'Promo2'])['Sales']
      .mean()
      .reset_index()
)

# Creating pivot table for grouped bar plotting
sales_pivot = promo2_sales.pivot(
    index='CompetitionAgeBucket',
    columns='Promo2',
    values='Sales'
)

# Renaming columns for clarity
sales_pivot.columns = ['No Promo2', 'Promo2 Active']

# Converting absolute sales into percentage values within each competition bucket
sales_percentage = sales_pivot.div(sales_pivot.sum(axis=1), axis=0) * 100

# --------------------------------------------------------------- #
# Starting the visualization
plt.figure(figsize=(12,6))

# Plotting grouped bar chart with percentage values
sales_percentage.plot(
    kind='bar',
    width=0.7,
    color=['#4C72B0', '#DD8452'],
    edgecolor='black',
    ax=plt.gca()
)

# Adding value labels on bars
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.1f%%', fontsize=9, padding=3)

# Adjusting x-tick labels for better readability
plt.xticks(rotation=0, ha='center')

# Adding chart title and axis labels
plt.title(
    "Sales Distribution (%) by Competition Age: Promo2 Active vs Non-Active",
    fontsize=16,
    fontweight='bold'
)
plt.xlabel("Competition Age (Months Since Competitor Opened)", fontsize=12)
plt.ylabel("Sales Percentage (%)", fontsize=12)

# Adjusting legend position for clarity
plt.legend(
    title="Promo2 Status",
    bbox_to_anchor=(0.5, -0.15),   # Moves legend below chart
    loc='upper center',
    ncol=2,
    frameon=True
)

# Adding horizontal gridlines only
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Improving layout
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* The chart shows that across all competition age buckets, sales share remains consistently higher on Non-Promo2 days compared to Promo2 Active days.

* The gap between Promo2 Active and Non-Promo2 sales percentage is remarkably stable, regardless of how long the competitor has been present.

* This indicates that long-running promotions (Promo2) are not effectively countering competitive pressure, even when competition is very new or extremely old.

###### 1. Why did you pick the specific chart?

* I picked this chart to understand whether the effectiveness of long-running promotions (Promo2) changes with competitive maturity.

* Using competition age buckets allows me to analyze if Promo2 performs better when competition is new versus well-established.

* A grouped bar chart with percentage values was chosen because it makes relative sales contribution immediately comparable across different competition stages, avoiding scale bias from absolute sales values.

###### 2. What is/are the insight(s) found from the chart?

* Promo2 consistently contributes less than 50% of sales across all competition age groups.

* The sales distribution gap (~5-8%) between Promo2 and Non-Promo2 remains almost unchanged as competition ages.

* This suggests Promo2 is structurally weak, not situationally weak — its underperformance is not driven by competition timing.

**⚠️ Negative-Growth Risks Identified -**

##### Chart 78 - Impact of Promo2 and Competition Age Sales Efficiency

````python
# Chart 78 - Visualisation Chart

# Features Used - Promo2, CompetitionOpenMonths, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart helps analyze how long-running promotions (Promo2) perform
# across different stages of competition maturity. By using SalesPerCustomer
# instead of total sales, we focus on customer value rather than footfall.
# It helps identify whether Promo2 sustains or loses effectiveness as
# competition becomes older.

# ---------------------------------------------------------------------------------------------------------- #

# Setting consistent visual style
sns.set(style='whitegrid')

# Creating competition age buckets using CompetitionOpenMonths
df['Competition_Age_Bucket'] = pd.cut(
    df['CompetitionOpenMonths'],
    bins=[-1, 6, 12, 24, 60, 120, float('inf')],
    labels=[
        'Very New (<6m)',
        'New (6–12m)',
        'Moderate (1–2y)',
        'Old (2–5y)',
        'Very Old (5–10y)',
        'Extremely Old (>10y)'
    ]
)

# Aggregating average SalesPerCustomer
promo2_comp_spc = (
    df.groupby(['Competition_Age_Bucket', 'Promo2'])['SalesPerCustomer']
      .mean()
      .reset_index()
)

# Creating the figure
plt.figure(figsize=(12, 6))

# Plotting line chart to observe trend across competition age
sns.lineplot(
    data=promo2_comp_spc,
    x='Competition_Age_Bucket',
    y='SalesPerCustomer',
    hue='Promo2',
    marker='o',
    palette=['#64B5F6', '#FFB74D']
)

# Adding chart title and labels
plt.title(
    'Average Sales per Customer by Competition Age: Promo2 Active vs Non-Active',
    fontsize=14,
    fontweight='bold'
)
plt.xlabel('Competition Age (Months Since Competitor Opened)')
plt.ylabel('Average Sales per Customer')

# Formatting legend
plt.legend(
    title='Promo2 Status',
    labels=['No Promo2', 'Promo2 Active'],
    loc='best'
)

# Rotating x-axis labels for readability
plt.xticks(rotation=20)

# Removing unnecessary borders
plt.box(False)

# Displaying the chart
plt.show()
````

##### Chart 81 - Average Sales per Customer by State Holiday: Promo2 Active vs Non-Active

````python
# Chart 81 - Promo2 × StateHoliday × Sales per Customer

# Features Used - Promo2, StateHoliday, Sales per Customer

# ---------------------------------------------------------------------------------------------------------- #
# ✅ Why this chart is important to include?
# This chart evaluates whether Promo2 improves spending efficiency (sales per customer)
# across different state holiday types.



# Calculate Sales per Customer safely
df_spc = df.copy()
df_spc['SalesPerCustomer'] = np.where(
    df_spc['Customers'] > 0,
    df_spc['Sales'] / df_spc['Customers'],
    0
)

# Aggregate data
spc_summary = (
    df_spc
    .groupby(['StateHoliday', 'Promo2'])['SalesPerCustomer']
    .mean()
    .reset_index()
)

# Plot setup
sns.set(style='whitegrid')
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=spc_summary,
    x='StateHoliday',
    y='SalesPerCustomer',
    hue='Promo2',
    palette=['#4C72B0', '#DD8452']
)

# Titles and labels
plt.title(
    'Average Sales per Customer by State Holiday: Promo2 Active vs Non-Active',
    fontsize=15,
    fontweight='bold'
)
plt.xlabel('State Holiday Type')
plt.ylabel('Sales per Customer')

# Legend customization
plt.legend(
    title='Promo2 Status',
    labels=['No Promo2', 'Promo2 Active'],
    loc='upper right'
)

# Add value labels on bars
for p in ax.patches:
    height = p.get_height()
    ax.annotate(
        f'{height:.2f}',
        (p.get_x() + p.get_width() / 2, height),
        ha='center',
        va='bottom',
        fontsize=9,
        xytext=(0, 3),
        textcoords='offset points'
    )

# Grid & aesthetics
plt.grid(axis='y', linestyle='--', alpha=0.4)
sns.despine()

# Adjusting layout
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* This chart analyzes how long-running promotions (Promo2) impact sales efficiency (Sales per Customer) across different state holiday types.

* Unlike total sales or customer count, this view isolates customer spending behavior, making it easier to assess whether Promo2 improves basket value rather than footfall.

* The chart clearly separates regular days (None) from state holidays (a, b, c), exposing a sharp contrast in customer purchasing patterns.

###### 1. Why did you pick the specific chart?

* This grouped bar chart is ideal for comparing two promotion states (Promo2 Active vs Non-Active) across multiple categorical holiday types.

* Using Sales per Customer removes the bias introduced by fluctuating customer volumes, which we already observed in earlier Promo2 analyses.

* This chart directly answers whether Promo2 changes how much a customer spends, not just how many customers visit.

###### 2. What is/are the insight(s) found from the chart?

* On non-holiday days, Promo2 significantly increases sales per customer, indicating higher basket value when long-running promotions are active.

* During state holidays (a, b, c), sales per customer drop sharply for both Promo2 and Non-Promo2, suggesting holiday shopping is need-based rather than promotion-driven.

* Promo2 does not compensate for reduced spending behavior during holidays, even though it slightly improves efficiency compared to non-promo in normal days.

##### Chart 83 - Sales Per Customer: Promo vs Promo2

````python
# Chart - 83 Visualisation Code

# Features Used: Promo, Promo2, SalesPerCustomer

# ✅ Why this chart is important to include?
# This chart is helping compare short-term promotions (Promo) and long-term promotions (Promo2)
# on a normalized efficiency metric — Sales Per Customer.
# By focusing on SalesPerCustomer instead of raw Sales or Customers,
# this chart is allowing a fair comparison of how each promotion strategy
# impacts customer-level revenue generation.

# ---------------------------------------------------------------------------------------------------------- #

# Preparing aggregated data for Promo and Promo2 comparison
promo_spc = (
    df.groupby(['Promo', 'Promo2'])['SalesPerCustomer']
      .mean()
      .reset_index()
)

# Converting Promo and Promo2 to readable labels
promo_spc['Promo'] = promo_spc['Promo'].map({0: 'No Promo', 1: 'Promo'})
promo_spc['Promo2'] = promo_spc['Promo2'].map({0: 'No Promo2', 1: 'Promo2'})

# Setting visualization style
sns.set(style='whitegrid')

# Creating the figure
plt.figure(figsize=(10,6))

# Creating grouped bar chart
sns.barplot(
    data=promo_spc,
    x='Promo',
    y='SalesPerCustomer',
    hue='Promo2',
    palette='Set2'
)

# Adding value labels on bars
for container in plt.gca().containers:
    plt.bar_label(
        container,
        fmt='%.2f',
        padding=3,
        fontsize=10
    )

# Adding titles and labels
plt.title(
    'Sales Per Customer: Promo vs Promo2',
    fontsize=16,
    fontweight='bold'
)
plt.xlabel('Daily Promotion (Promo)', fontsize=12)
plt.ylabel('Sales Per Customer', fontsize=12)

# Adjusting legend placement
plt.legend(
    title='Long-term Promotion (Promo2)',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

# Improving layout and grid
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

# Displaying the plot
plt.show()
````

**Summary -**

* This chart compares the combined and individual impact of short-term promotions (Promo) and long-term promotions (Promo2) on Sales Per Customer (SPC).

* It clearly shows that Promo and Promo2 are complementary rather than redundant, and when both are active together, stores achieve the highest customer-level revenue efficiency.

###### 1. Why did you pick the specific chart?

This grouped bar chart was chosen because:

* It enables direct comparison between:

    * Promo only

    * Promo2 only

    * Promo + Promo2 combined

* Sales Per Customer is a normalized metric, removing bias from store size or footfall.

* The grouped bar format makes it easy to visually isolate incremental gains from Promo2 when Promo is already running.

* This chart acts as a decision-level visualization that helps answer whether running both promotions together makes sense.

###### 2. What is/are the insight(s) found from the chart?

Insight 1 — Promo increases customer-level spending

* Without Promo, SPC is ~ 6.37

* With Promo, SPC jumps to ~ 9.37 ➡️ Daily promotions significantly increase average spend per customer.

Insight 2 — Promo2 adds incremental value

* Without Promo, Promo2 still increases SPC from 6.37 → 6.92 ➡️ Promo2 independently improves spending behavior, even without daily discounts.

Insight 3 — Best performance occurs when Promo & Promo2 run together

* Promo + Promo2 yields the highest SPC (~10.40) ➡️ Long-term promotions amplify the effectiveness of short-term promotions rather than cannibalizing them.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**✅ Positive Business Impact :**

Yes, absolutely:

* It removes the confusion caused by grouped bar charts and highlights a critical structural pattern: Assortment b is exclusively associated with StoreType b, while other store types operate only with assortments a and c.

* This chart validates business logic and prevents misleading cross-comparisons.

###### 1. Why did you pick the specific chart?

* A heatmap is ideal when:

    * Two categorical variables interact (StoreType × Assortment)

    * Some combinations do not exist

* Unlike bar charts, the heatmap:

    * Clearly shows missing combinations as blank/NaN

    * Prevents false assumptions (e.g., comparing StoreType a with Assortment b)

* It also makes relative sales intensity immediately visible through color gradients.

##### Chart 86 - Promo × Day × Month × Sales Interaction

````python
# Chart - 86 Visualisation Code

# Features Used - Promo, Sales, DayOfWeek, Month

# ✅ Why this chart is important to include?
# This chart is chosen to analyze the combined interaction of promotion, weekday behavior,
# and monthly seasonality on sales.
# A faceted heatmap allows us to compare Promo vs No Promo side-by-side  while simultaneously
# observing: Day-of-week patterns, Month-wise seasonality, Intensity of average sales.
# This approach avoids visual clutter and makes a 4-variable relationship interpretable.
# ---------------------------------------------------------------------------------------------------------- #

# Creating a pivot table for average sales
heatmap_data = df.groupby(['Promo', 'Month', 'DayOfWeek'])['Sales'].mean().reset_index()

# Creating two subsets for Promo and No Promo
promo_0 = heatmap_data[heatmap_data['Promo'] == 0]
promo_1 = heatmap_data[heatmap_data['Promo'] == 1]

# Pivoting data for heatmap structure
pivot_no_promo = promo_0.pivot(index='Month', columns='DayOfWeek', values='Sales')
pivot_promo = promo_1.pivot(index='Month', columns='DayOfWeek', values='Sales')

# Setting plot style
sns.set(style='white')

# Creating subplots
fig, axes = plt.subplots(1, 2, figsize=(18, 8), sharey=True)

# Heatmap for No Promo
sns.heatmap(
    pivot_no_promo,
    ax=axes[0],
    cmap='Blues',
    annot=False,
    fmt='.0f',
    cbar=True
)
axes[0].set_title('Average Sales (No Promo)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Day of Week')
axes[0].set_ylabel('Month')

# Heatmap for Promo
sns.heatmap(
    pivot_promo,
    ax=axes[1],
    cmap='Oranges',
    annot=False,
    fmt='.0f',
    cbar=True
)
axes[1].set_title('Average Sales (Promo Active)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Day of Week')
axes[1].set_ylabel('')

# Final layout adjustment
plt.suptitle(
    'Promo × Sales × Day × Month Interaction',
    fontsize=16,
    fontweight='bold'
)

# Displaying the plot
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

**Summary -**

* This faceted heatmap clearly highlights how promotions change sales behavior across both weekdays and months.

* Promo Active days consistently generate higher sales across almost all months and weekdays.

* Month-level seasonality is strong: December (Month 12) shows the highest sales intensity, especially when promotions are active.

* Weekday patterns differ with and without promo:
Without promo, sales are more evenly spread across weekdays.

* With promo, early weekdays (Day 1–3) and mid-week days benefit the most.

* This chart successfully uncovers hidden temporal interactions that single-variable charts cannot reveal.

###### 1. Why did you pick the specific chart?

* I picked a faceted heatmap because, This analysis involves four dimensions simultaneously: Promo status,  Month, Day of week, Sales

* Heatmaps are ideal for spotting intensity patterns, seasonality, and interaction effects.

* Faceting Promo vs No Promo avoids overcrowding while enabling direct visual comparison.

* A simple bar or line chart would fail to capture this multi-level interaction clearly.

###### 2. What is/are the insight(s) found from the chart?

**🔹 Key Insights**

* Promotions amplify sales across all months
Promo-active heatmap consistently shows warmer colors than no-promo.

* Confirms that promotions increase revenue beyond normal seasonality.

* December (Month 12) is the strongest month

* Highest sales intensity appears in Month 12, especially with promo.

* Suggests strong year-end festive demand.

* Promo effectiveness varies by weekday

* Promo impact is strongest on Day 1–3 (early weekdays).

* Weekends or late-week days show relatively lower incremental gains.

* Without promo, sales are flatter

* No-promo sales rely mostly on seasonality, not weekday effects.

* Indicates promotions introduce behavioral demand shifts, not just volume increases.

Are there any insights that lead to negative growth? Justify with specific reason.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

* Based on the extensive univariate, bivariate, and multivariate chart experiments performed in the Exploratory Data Analysis phase, I have observed several strong patterns and relationships between sales, customers, promotions, and store characteristics.

* To statistically validate whether these observed patterns are truly significant or occurred due to random variation, hypothesis testing is required.

* Accordingly, I have formulated three hypothesis statements based on the most impactful and business-critical insights derived from the visual analysis.

* Each hypothesis focuses on a key business question related to promotions, customer behavior, and sales efficiency.

* In the following sections, appropriate statistical tests will be applied to each hypothesis to test the assumptions made during EDA and to draw data-driven conclusions with statistical confidence.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

There is a statistically significant difference in the average daily sales between promotional days (Promo = 1) and non-promotional days (Promo = 0).

Let,

μ₁ = Mean daily sales on promotional days

μ₀ = Mean daily sales on non-promotional days


**Null Hypothesis (H₀):**

There is no statistically significant difference in the average daily sales between promotional days (Promo = 1) and non-promotional days (Promo = 0).


H₀ : μ₀ = μ₁


**Alternative Hypothesis (H₁):**

There is a statistically significant difference in the average daily sales between promotional days (Promo = 1) and non-promotional days (Promo = 0).


H₁ : μ₀ ≠ μ₁


#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# ---------------------------------------------------------------------------------------------------------- #

# Selecting only open store days to avoid bias from closed days
promo_sales = df[(df['Promo'] == 1) & (df['Open'] == 1)]['Sales']
non_promo_sales = df[(df['Promo'] == 0) & (df['Open'] == 1)]['Sales']

# Importing required statistical function
from scipy.stats import ttest_ind

# Performing Welch’s t-test (does not assume equal variance)
t_statistic, p_value = ttest_ind(
    promo_sales,
    non_promo_sales,
    equal_var=False
)

# Displaying test results
print("Independent Two-Sample t-test (Welch’s Test) Results")
print(f"T-statistic : {t_statistic:.4f}")
print(f"P-value     : {p_value:.6f}")
````

##### Which statistical test have you done to obtain P-Value?

* To evaluate whether there is a statistically significant difference in average sales between promotional and non-promotional days, an Independent Two-Sample t-test (Welch's t-test) was performed.

* This test compares the mean sales of two independent groups — days with promotion (Promo = 1) and days without promotion (Promo = 0).

* The test calculates a t-statistic based on the difference between the sample means, adjusted for variability and sample size, and then derives the corresponding p-value to assess statistical significance.

* The resulting p-value was used to determine whether the observed difference in mean sales is statistically significant or could have occurred due to random chance.



##### Why did you choose the specific statistical test?

* I chose the Independent Two-Sample t-test (Welch's t-test) because the objective was to compare the mean sales values of two independent groups — promotional days and non-promotional days.

* The two groups are independent in nature, as daily sales recorded on promotional days do not influence sales on non-promotional days.

* The population variance for sales is unknown, and the observed variability in sales between promo and non-promo days is not guaranteed to be equal.

* Welch's t-test is well suited for such situations because it does not assume equal variances between the two groups.

* Additionally, the dataset contains a large number of observations, making Welch's t-test robust even if the underlying sales distribution is not perfectly normal.

* Therefore, Welch's t-test provides a reliable and statistically sound method to determine whether the observed difference in mean sales is significant.

**📌 Conclusion -**

Decision Rule

* Significance level (α) = 0.05

* Obtained p-value = 0.000000

Since

p-value < α

**Statistical Conclusion -**

* The null hypothesis (**H₀**) is **rejected**.

* There is **strong statistical evidence** to suggest that the mean daily sales on promotional days are significantly different from the mean daily sales on non-promotional days.

**Business Interpretation -**

* Promotional days generate statistically higher daily sales compared to non-promotional days.

* The extremely large t-statistic (356.64) indicates that the observed difference is not due to random chance, but represents a real and consistent effect.

* This confirms the patterns observed during EDA and validates the effectiveness of short-term promotions as a key revenue driver.


**Final Statement -**

* Based on the Welch's two-sample t-test results, the **null hypothesis is rejected at a 5% significance level**.

* This indicates that promotions have a statistically significant impact on daily sales, confirming that promotional activities play a crucial role in driving higher revenue for Rossmann stores.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

There is a statistically significant difference in Sales Per Customer between days when Promo2 is active and when it is not active.


Let,

μ₀ : SalesPerCustomer when promo2 is not active. (IsPromo2Active = 0)


μ₁ : SalesPerCustomer when promo2 is active. (IsPromo2Active = 1)


**Null Hypothesis (H₀) :**

Long-running promotions do not change the average sales per customer.

H₀ : μ₀ = μ₁


**Alternate Hypothesis (H₁) :**

Long-running promotions significantly affect the average sales per customer.

H₁ : μ₀ ≠ μ₁

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Performing statistical test for Hypothesis Statement 2

# Comparing SalesPerCustomer between Promo2 Active and Non-Active days

from scipy import stats

# Separating the two independent samples
spc_promo2_active = df[df['IsPromo2Active'] == 1]['SalesPerCustomer']
spc_promo2_inactive = df[df['IsPromo2Active'] == 0]['SalesPerCustomer']


# Performing Welch’s Two-Sample t-test
t_statistic, p_value = stats.ttest_ind(
    spc_promo2_active,
    spc_promo2_inactive,
    equal_var=False
)

# Displaying the results
print("Welch’s Two-Sample t-test Results")
print(f"T-statistic : {t_statistic:.4f}")
print(f"P-value     : {p_value:.6f}")
````

##### Which statistical test have you done to obtain P-Value?

* An Independent Two-Sample t-test (Welch’s t-test) was performed.

* This test compared the mean Sales Per Customer between two independent groups:

  * Days when Promo2 was active

  * Days when Promo2 was not active

##### Why did you choose the specific statistical test?

* The objective was to compare the mean Sales Per Customer across two independent conditions.

* Sales Per Customer is a continuous numeric variable, making it suitable for a mean comparison test.

* The variance and sample size between Promo2-active and non-active days are not guaranteed to be equal.

* Welch's t-test is robust to unequal variances and large real-world datasets, making it appropriate for this analysis.

**🔹 Interpretation of Statistical Test Output -**

* The t-statistic (44.33) is extremely large, which means the difference in Sales Per Customer between Promo2-active and Promo2-inactive days is far larger than what random variation would produce.

* The p-value is effectively 0, which is much smaller than the significance level (α = 0.05).
This provides very strong statistical evidence against the null hypothesis.


**📈 Business Interpretation -**

**✅ Positive Business Impact :**

* This result confirms that long-running promotions (Promo2) materially affect sales efficiency.

* The company can no longer assume Promo2 improves revenue quality simply by increasing customer volume.

* Sales Per Customer is a strong profitability metric, and this finding allows:

    * Better evaluation of long-term promotion ROI

    * Redesigning Promo2 strategies to focus on higher-value customers

    * Store-type–specific or assortment-based Promo2 targeting

**⚠️ Negative Growth Risk -**

* The strong statistical difference suggests that Promo2 may be attracting lower-spending customers.

* If Promo2 continues without redesign:

  * Revenue growth may stagnate despite higher footfall

  * Profit margins may shrink due to discount-heavy behavior

  * Stores may become dependent on promotions without improving efficiency

* This creates a long-term revenue dilution risk if Promo2 is not optimized.


**Final Conclusion -**

**Decision: Reject the Null Hypothesis (H₀)**

* Since the p-value is significantly smaller than 0.05, the null hypothesis is rejected.

* There is a statistically significant difference in Sales Per Customer between Promo2-active and Promo2-inactive days.


### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Customer footfall varies significantly across different days of the week, indicating that customer shopping behavior is not uniform throughout the week.


**Null Hypothesis (H₀):**

There is no statistically significant difference in average customer footfall across different days of the week.

H₀: μ₁ = μ₂ = μ₃ = μ₄ = μ₅ = μ₆ = μ₇


Where:
μ₁, μ₂, …, μ₇ :  represent the mean customer footfall for each day of the week.


**Alternate Hypothesis (H₁):**

At least one day of the week has a statistically significant difference in average customer footfall compared to the others.

H₁: At least one μᵢ differs from the others

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Performing statistical test for Hypothesis Statement 3

# Comparing average customer footfall across different days of the week

# Creating a list of customer samples for each day of the week
customers_by_day = [
    df[df['DayOfWeek'] == day]['Customers']
    for day in sorted(df['DayOfWeek'].unique())
]

# Performing One-Way ANOVA
f_statistic, p_value = stats.f_oneway(*customers_by_day)

# Displaying the results
print("One-Way ANOVA Results")
print(f"F-statistic : {f_statistic:.4f}")
print(f"P-value     : {p_value:.6f}")
````

##### Which statistical test have you done to obtain P-Value?

* A One-Way Analysis of Variance (ANOVA) test was performed.

* This test was used to compare the mean customer footfall across multiple groups, where each group represents a different day of the week.

##### Why did you choose the specific statistical test?

* The objective was to compare average customer footfall across more than two categories (seven weekdays).

* Customer count is a continuous numeric variable, while DayOfWeek is a categorical variable with multiple levels.

* A two-sample t-test is not suitable in this scenario because it can only compare two groups at a time.

* One-Way ANOVA is specifically designed to test whether at least one group mean differs significantly when comparing multiple independent groups.

**🔹 Interpretation of Statistical Test Output -**

* The F-statistic is extremely large, which indicates that the variation in average customer footfall between different weekdays is far greater than the variation within the same weekday.

* The p-value is effectively 0, which is much smaller than the significance level (α = 0.05).
This provides very strong statistical evidence against the null hypothesis.


**Conclusion -**

**Decision: Reject the Null Hypothesis (H₀)**

* Since the p-value is significantly smaller than 0.05, the null hypothesis is rejected.

* There is a statistically significant difference in average customer footfall across different days of the week.




**📈 Business Interpretation -**

**✅ Positive Business Impact :**

* This result confirms that customer shopping behavior varies strongly by day of the week.

* The business can use this insight to:

  * Optimize staff scheduling on high-footfall days

  * Improve inventory planning and replenishment

  * Run targeted promotions on low-footfall weekdays

  * Enhance customer experience during peak days

* Understanding daily traffic patterns helps improve both revenue generation and cost efficiency.

**⚠️ Negative Growth Risks -**

* If weekday differences in footfall are ignored:

  * Stores may be understaffed on peak days, leading to lost sales and poor service

  * Resources may be wasted on low-traffic days, increasing operational costs

  * Promotional efforts may be misaligned with actual customer behavior

* This could result in inefficient operations and missed revenue opportunities over time.

### 3. Categorical Encoding

````python
# Encode your categorical columns


# ---------------------------------------------------------
# CATEGORICAL ENCODING
# ---------------------------------------------------------

from sklearn.preprocessing import LabelEncoder

# Creating a copy to preserve original dataset if needed
df_encoded = df.copy()

# -----------------------------
# 1. Encoding StateHoliday
# -----------------------------
# StateHoliday is a nominal categorical feature with low cardinality
stateholiday_encoder = LabelEncoder()
df_encoded['StateHoliday_Encoded'] = stateholiday_encoder.fit_transform(df_encoded['StateHoliday'])

# -----------------------------
# 2. Encoding StoreType
# -----------------------------
# StoreType is a nominal categorical feature
storetype_encoder = LabelEncoder()
df_encoded['StoreType_Encoded'] = storetype_encoder.fit_transform(df_encoded['StoreType'])

# -----------------------------
# 3. Encoding Assortment (Ordinal)
# -----------------------------
# Assortment has an inherent order: a < b < c
assortment_mapping = {'a': 0, 'b': 1, 'c': 2}
df_encoded['Assortment_Encoded'] = df_encoded['Assortment'].map(assortment_mapping)


# -----------------------------
# 4. Encoding PromoInterval
# -----------------------------
# PromoInterval is categorical with multiple combinations
promointerval_encoder = LabelEncoder()
df_encoded['PromoInterval_Encoded'] = promointerval_encoder.fit_transform(df_encoded['PromoInterval'])



# -----------------------------
# Dropping original categorical columns
# -----------------------------
df_encoded.drop(
    columns=['StateHoliday', 'StoreType', 'Assortment', 'PromoInterval'],
    inplace=True
)

print("✅ Categorical Encoding Completed Successfully!")
display(df_encoded[['StateHoliday_Encoded', 'StoreType_Encoded',
                     'Assortment_Encoded', 'PromoInterval_Encoded']].head())


# ---------------------------------------------------------
# Dropping EDA-only bucket columns before modeling
# ---------------------------------------------------------

bucket_columns = [
    'CompetitionDistanceBin',
    'CompetitionDistanceBucket',
    'CompetitionAgeBucket',
    'Competition_Age_Bucket',
    'MonthName'
]

df_encoded.drop(columns=bucket_columns, errors='ignore', inplace=True)

print("✅ EDA-only bucket columns dropped before modeling!")



````

#### What all categorical encoding techniques have you used & why did you use those techniques?

* During the feature engineering stage, categorical encoding was performed to convert non-numeric variables into a machine-learning-friendly format.

* Label encoding was used for nominal categorical features such as StateHoliday, StoreType, and PromoInterval.

* This approach was chosen because these variables have a limited number of categories and the dataset size is very large. Label encoding avoids unnecessary feature expansion and keeps the feature space compact and computationally efficient.

* For the Assortment feature, ordinal encoding was applied instead of one-hot encoding.

* This decision was based on the business meaning of the variable, as assortment types follow a natural order from basic to extended product ranges.

* Preserving this order allows the model to better capture differences in product variety.

* One-hot encoding was intentionally avoided at this stage to prevent high dimensionality and sparsity, especially given the large number of records in the dataset.

* Tree-based machine learning models can effectively work with label-encoded categorical features without assuming linear relationships, making label encoding a suitable choice.

* Overall, the selected encoding techniques strike a balance between interpretability, computational efficiency, and model compatibility, ensuring that the dataset remains scalable and ready for further preprocessing and modeling.

### 5. Feature Manipulation & Selection

#### 1. Feature Manipulation

````python
# Manipulate Features to minimize feature correlation and create new features

# ===============================
# FEATURE MANIPULATION
# ===============================
# Purpose:
# - Reducing feature redundancy
# - Handling cyclic temporal patterns
# - Stabilizing skewed numeric features
# ===============================

# Creating a copy to preserve original dataframe
df_feat = df_encoded.copy()

# ------------------------------------------------
# Dropping raw date column (time already extracted)
# ------------------------------------------------
df_feat.drop(columns=['Date'], inplace=True)

# ------------------------------------------------
# Creating cyclic features for Month
# ------------------------------------------------
df_feat['Month_sin'] = np.sin(2 * np.pi * df_feat['Month'] / 12)
df_feat['Month_cos'] = np.cos(2 * np.pi * df_feat['Month'] / 12)

# ------------------------------------------------
# Creating cyclic features for DayOfWeek
# ------------------------------------------------
df_feat['DayOfWeek_sin'] = np.sin(2 * np.pi * df_feat['DayOfWeek'] / 7)
df_feat['DayOfWeek_cos'] = np.cos(2 * np.pi * df_feat['DayOfWeek'] / 7)

# ------------------------------------------------
# Log-transforming CompetitionDistance to reduce skew
# ------------------------------------------------
df_feat['CompetitionDistance_log'] = np.log1p(df_feat['CompetitionDistance'])

print("✅ Feature manipulation completed successfully")

print("✅ Feature manipulation completed with business-aligned transformations.")
````

#### 2. Feature Selection

````python
# Select your features wisely to avoid overfitting

# ===============================
# FEATURE SELECTION
# ===============================
# Purpose:
# - Removing target leakage
# - Removing redundant features
# - Keeping only prediction-safe inputs
# ===============================

df_model = df_feat.copy()

# Dropping unnecessary and leakage-prone columns
df_model.drop(columns=[
    'Store',
    'Month',
    'DayOfWeek',
    'Promo2',
    'Customers',
    'SalesPerCustomer',
    'ZeroSalesWhileOpen',
    'CompetitionDistance',
    'CompetitionDistance_NA',
    'CompetitionOpenSinceMonth',
    'CompetitionOpenSinceYear',
    'CompetitionOpenSince_NA',
    'Promo2SinceWeek',
    'Promo2Since_NA',
    'PromoInterval_Encoded'
], inplace=True)

print("✅ Feature selection completed successfully")
print("Final columns used for modeling:")
print(df_model.columns.tolist())
````

##### What all feature selection methods have you used  and why?

* Feature selection was primarily performed using domain knowledge - driven selection combined with correlation and redundancy analysis rather than automated statistical elimination techniques.

* First, business understanding and real-world forecasting constraints were applied to identify features that would be available at prediction time.

* Variables such as Customers, SalesPerCustomer, and ZeroSalesWhileOpen were excluded because they contain post-event information or are directly derived from the target variable, which would otherwise introduce target leakage.

* Second, redundant and highly correlated features were removed to minimize multicollinearity. For example, raw temporal features (Month, DayOfWeek) were replaced by their cyclic representations (Month_sin, Month_cos, DayOfWeek_sin, DayOfWeek_cos) to better capture seasonality while avoiding duplicate information.

* Third, structurally redundant competition features were eliminated.

* Instead of using raw competition opening dates and missing value flags, consolidated features such as CompetitionDistance_log and CompetitionOpenMonths were retained, as they more effectively represent competitive pressure while reducing feature complexity.

* Encoded categorical variables (StoreType_Encoded, Assortment_Encoded, StateHoliday_Encoded) were retained based on their strong explanatory power observed during exploratory analysis.

* Overall, this hybrid approach ensured that the final feature set is leakage-free, non-redundant, interpretable, and aligned with real-world deployment scenarios, making it suitable for robust and generalizable model training.

##### Which all features you found important and why?

* The most important features identified in this project are those that directly influence daily sales while being available before the day of sale, ensuring realistic and deployable predictions.

* **Promotion-related features** (Promo, IsPromo2Active) were found to be highly important, as extensive EDA showed that both short-term and long-running promotions significantly impact sales, customer behavior, and sales efficiency. These variables capture marketing-driven demand fluctuations.

* **Temporal features** (Year, Month_sin, Month_cos, DayOfWeek_sin, DayOfWeek_cos) emerged as critical predictors because sales exhibit strong seasonality and weekly patterns. Cyclic encoding was used to preserve the natural continuity of time (e.g., December to January, Sunday to Monday).

* **Store characteristics** (StoreType_Encoded, Assortment_Encoded) were identified as important since different store formats and assortment depths consistently showed varying sales performance and customer engagement during EDA.

* **Competition-related features** (CompetitionDistance_log, CompetitionOpenMonths) were retained because they effectively represent the intensity and maturity of nearby competition. Log transformation of competition distance reduced skewness and improved its usability as a predictive signal.

* **Holiday indicators** (SchoolHoliday, StateHoliday_Encoded) were also found to be important, as sales patterns differed noticeably between holiday and non-holiday periods, especially when combined with promotions.

* Features such as Customers, SalesPerCustomer, and ZeroSalesWhileOpen were intentionally excluded despite their strong correlation with sales, as they contain post-event information and would cause target leakage if used for prediction.

* Overall, the selected features balance **business relevance, statistical usefulness, and modeling integrity**, ensuring that the model learns from genuine demand drivers rather than artifacts of the target variable

### ML Model - 2

````python
# ML Model - 2 Implementation


# ================================
# MODEL 1 - LINEAR REGRESSION
# ================================

# Features Used:
# All selected numerical and encoded features
# Target Used: Log-Transformed Sales

# ✅ Why this model is important to include?
# This model is serving as the first learnable machine learning model after baseline.
# Linear Regression is helping to establish a strong benchmark and understand
# whether linear relationships exist between engineered features and sales.
# Using log-transformed sales is stabilizing variance and reducing skewness,
# making the model assumptions more valid.

# --------------------------------------------------------------- #

# Importing required model and evaluation libraries
from sklearn.linear_model import LinearRegression


# --------------------------------------------------------------- #
# Initializing Linear Regression model
lr_model = LinearRegression()

# --------------------------------------------------------------- #
# Fit the Algorithm
lr_model.fit(x_train, y_train)

# Predict on the model
# Making predictions on test data
y_pred = lr_model.predict(x_test)

# --------------------------------------------------------------- #
# Evaluating model performance

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

# Printing evaluation results
print("📌 Linear Regression Model Performance (Log-Transformed Sales)")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart


# Extracting the data for the plot
metrics = ['MAE', 'RMSE', 'R²']
values = [mae, rmse, r2]


# Creating the bar chart
plt.figure(figsize=(8,5))
sns.barplot(
    x=metrics,
    y=values,
    palette='Blues'
)

# Setting title and labels
plt.title(
    'Evaluation Metrics for Linear Regression Model',
    fontsize=16,
    fontweight='bold'
)
plt.ylabel('Metric Value')
plt.xlabel('Metric Type')
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Displaying the plot
plt.tight_layout()
plt.show()
````

* I have used **Linear Regression** as the first learnable machine learning model after the baseline model.

* Linear Regression helps in understanding whether a **linear relationship exists between the engineered features and sales**, making it a strong benchmark model.

* The model was trained using **log-transformed Sales** as the target variable to reduce skewness and stabilize variance.

* From the evaluation metric score chart:

  * The model achieves **very low MAE and RMSE values**, indicating small prediction errors.

  * The **R² score of 0.9887** shows that the model explains **nearly 99% of the variance** in sales.

* Overall, the evaluation chart confirms that Linear Regression performs **exceptionally well** on the processed dataset.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)


# Cross-Validation and Hyperparameter Tuning using GridSearchCV

param_grid = {
    'fit_intercept': [True, False]
}

grid_search = GridSearchCV(
    estimator=LinearRegression(),
    param_grid=param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)


# Fit the Algorithm
# Fitting GridSearch on training data
grid_search.fit(x_train, y_train)


# Predict on the model
# Fetching best model
best_lr_model = grid_search.best_estimator_

# Predicting with best model
y_pred_best = best_lr_model.predict(x_test)


# Evaluating best model performance
mae_best = mean_absolute_error(y_test, y_pred_best)
rmse_best = np.sqrt(mean_squared_error(y_test, y_pred_best))
r2_best = r2_score(y_test, y_pred_best)


# Printing optimized model performance
print("\n📌 Optimized Linear Regression Model Performance")
print(f"MAE  : {mae_best:.4f}")
print(f"RMSE : {rmse_best:.4f}")
print(f"R²   : {r2_best:.4f}")
````

* I have applied **5-fold Cross-Validation** using GridSearchCV.

* Cross-validation ensures that the model performance is **stable across different subsets of the data** and not dependent on a single train-test split.

* Hyperparameter tuning was performed to test different configurations of the Linear Regression model.

* GridSearchCV was chosen because:

  * An exhaustive grid search is computationally efficient and easy to interpret.

  * It ensures that the **best possible parameter combination** is tested systematically.

* For this model, the parameter fit_intercept was tuned. Here.

### ML Model - 3

````python
# ML Model - 3 Implementation


# 📌 Model 3 - Random Forest Regression

# Features Used - All selected input features
# Target Variable - Log-Transformed Sales

# ✅ Why this model is important to include?
# This model is helping to capture non-linear relationships between features and sales.
# Unlike Linear Regression, Random Forest does not assume linearity and can automatically
# learn complex interactions between promotions, competition, store characteristics,
# and customer behavior.
# This model is also useful to validate whether the strong performance of the linear model
# is genuine or influenced by linear correlations.

# ---------------------------------------------------------------------------------------------------------- #

# Importing required library
from sklearn.ensemble import RandomForestRegressor


# Initializing the Random Forest Regressor
rf_model = RandomForestRegressor(
       n_estimators=100,
    max_depth=15,          # limiting tree depth
    min_samples_leaf=50,   # prevents overfitting
    random_state=42,
    n_jobs=-1
)

# Fit the Algorithm
# Fitting the model on training data
rf_model.fit(x_train, y_train)


# Predict on the model
# Predicting on test data
y_pred_rf = rf_model.predict(x_test)

# Evaluating model performance
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# Displaying evaluation results
print("📌 Random Forest Model Performance (Log-Transformed Sales)")
print(f"MAE  : {mae_rf:.4f}")
print(f"RMSE : {rmse_rf:.4f}")
print(f"R²   : {r2_rf:.4f}")
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart

# Visualizing evaluation metric score chart for Random Forest

# Creating Data for the plot
metrics = ['MAE', 'RMSE', 'R²']
scores = [mae_rf, rmse_rf, r2_rf]

# Setting figure size
plt.figure(figsize=(8,5))

# Creating bar chart
sns.barplot(x=metrics, y=scores, palette='viridis')

# Showing values on bars
for idx, value in enumerate(scores):
    plt.text(idx, value, f"{value:.3f}", ha='center', va='bottom', fontsize=11)

# Setting title and labels
plt.title("Random Forest Evaluation Metrics", fontsize=15, fontweight='bold')
plt.ylabel("Metric Value")
plt.xlabel("Evaluation Metric")

# Displaying the plot
plt.grid(axis='y', alpha=0.3)
plt.show()
````

* I have used a **Random Forest Regressor**, which is an ensemble-based machine learning model that combines predictions from multiple decision trees.

* The model works by training each tree on a random subset of data and features and then averaging their predictions, which helps reduce variance and improve generalization.

* Random Forest is particularly suitable for this problem because sales are influenced by multiple interacting factors such as promotions, competition, store type, and customer behavior, which often show non-linear relationships.

* The model was trained using **log-transformed sales** as the target variable to stabilize variance and reduce the impact of extreme sales values.

* Model performance was evaluated using **MAE, RMSE, and R²**, and the results were visualized using an evaluation metric score chart.

**Model Performance Interpretation**

* **MAE (0.1880)** indicates that, on average, the model’s predictions deviate from the actual log-sales values by a very small margin.

* **RMSE (0.2789)** shows that large prediction errors are well controlled, meaning the model performs consistently even on high-sales days.

* **R² (0.9930)** suggests that the model explains 99.3% of the variance in sales, which indicates an excellent fit and strong predictive capability.

**What the Evaluation Metric Score Chart Shows**

* The evaluation metric score chart visually confirms the strong performance of the Random Forest model.

* Low error metrics (MAE and RMSE) combined with a very high R² value demonstrate that the model captures complex sales patterns effectively.

* This chart helps compare performance across models later and provides a clear, interpretable summary of model accuracy for business stakeholders.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Importing necessar libraries
from sklearn.model_selection import RandomizedSearchCV

# Creating a stratified sample for hyperparameter tuning
df_sample = df_scaled.sample(300_000, random_state=42)

x_sample = df_sample[x_train.columns]
y_sample = np.log1p(df_sample['Sales'])

# Creating parameter grid for RandomSearchCV
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [10, 15, 20],
    'min_samples_leaf': [20, 50, 100]
}

# Fit the Algorithm

# Initializing the model
rf = RandomForestRegressor(random_state=42, n_jobs=-1)

# Otimizing the model by cross -validating and hyper -parameter tuning
rf_random = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter=3,
    cv=2,
    scoring='r2',
    n_jobs = -1,
    verbose=1,
    random_state=42
)

# Fitting the model
rf_random.fit(x_sample, y_sample)

# Finding the best model
best_rf = rf_random.best_estimator_

# Predict on the model
# Predicting on test data using best estimator from RandomizedSearchCV
y_pred_best_rf = best_rf.predict(x_test)

# Calculating evaluation metrics
mae_best_rf = mean_absolute_error(y_test, y_pred_best_rf)
rmse_best_rf = np.sqrt(mean_squared_error(y_test, y_pred_best_rf))
r2_best_rf = r2_score(y_test, y_pred_best_rf)

# Displaying optimized model performance
print("📌 Optimized Random Forest Performance (After Hyperparameter Tuning)")
print(f"MAE  : {mae_best_rf:.4f}")
print(f"RMSE : {rmse_best_rf:.4f}")
print(f"R²   : {r2_best_rf:.4f}")
````

##### Which hyperparameter optimization technique have you used and why?

* I have used **Randomized Search Cross-Validation (RandomizedSearchCV)** for hyperparameter optimization.

* This technique randomly samples a fixed number of hyperparameter combinations from a predefined search space instead of evaluating all possible combinations.

* Given the **large size of the dataset**, using exhaustive methods like GridSearchCV would be computationally expensive and time-consuming.

* RandomizedSearchCV allows efficient exploration of the most impactful hyperparameters such as the number of trees, tree depth, and minimum samples per leaf.

* To further reduce execution time while maintaining model reliability, cross-validation was performed using **2 folds** on a **representative sample of the dataset.**

* The optimized model shows **slightly improved performance** (lower MAE and RMSE with a marginal increase in R²), confirming that the chosen optimization strategy was effective.

* This approach provides a strong balance between **model performance, computational efficiency, and practical feasibility**, making it suitable for real-world deployment.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

* Yes, a measurable improvement was observed after applying cross-validation and hyperparameter tuning using RandomizedSearchCV.

* The initial Random Forest model was trained with manually selected parameters, whereas the optimized model used systematically tuned hyperparameters obtained through cross-validation.

* After optimization, the model demonstrated **lower error metrics** and a **slight increase in explanatory power**, confirming that the tuning process was effective.
